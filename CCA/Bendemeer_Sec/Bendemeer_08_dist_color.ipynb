```python
from ugot import ugot
got = ugot.UGOT()
got.initialize("192.168.88.1")

IR = got.read_distance_data(21) # 0 to 250

# move forward until there is an obstacle in front
while IR > 50:
    got.mecanum_move_speed(0, 30)
    IR = got.read_distance_data(21)
    print(IR)
got.mecanum_stop()
```

The following cell uses a `while` loop to continuously update the `IR` variable with the current distance, and then checks if the distance to an obstacle is too far, too near, or within the accepted range. 
- If the obstacle is too far, the robot will move forward. 
- If the obstacle is too near, the robot will move backwards. 
- If the obstacle is within the range, the robot will stop moving.

```python
# keep something at a constant distance
while True:
    IR = got.read_distance_data(21)
    if IR > 50:
        # move forward
        got.mecanum_move_speed(0, 30)
    elif IR < 30:
        # move backward
        got.mecanum_move_speed(1, 30)
    else:
        # stop
        got.mecanum_stop()
```

Using the same logic, we can code the robot to keep a ball centered in the camera view. The explanation of the coordinate system is at [this link](https://docs.ubtrobot.com/ugot/#/en-us/extension/Basic_Building_Blocks/AI_visual/Overall).
- If the ball is too far left, the robot will shift left. 
- If the ball is too far right, the robot will shift right. 
- If the ball is exactly in the centre, the robot will stop moving.

```python
from ugot import ugot
got = ugot.UGOT()
got.initialize("192.168.88.1")

got.load_models(["color_recognition"])
info = got.get_color_total_info()
print(info) # [color, shape, center_x, center_y, height, width, area]
while True:
    info = got.get_color_total_info()
    if info: # check that it recognises a ball
        x_coord = info[2] # get the 3rd item in the list
        print(x_coord)
        if x_coord < 320: # ball is on the left
            got.mecanum_move_xyz(-5, 0, 0) # shift left
        elif x_coord > 320: # ball is on the right
            got.mecanum_move_xyz(5, 0, 0) # shift right
        else:
            got.mecanum_stop() # stop
```
