```
from ugot import ugot
got = ugot.UGOT()
got.initialize("192.168.88.1")

print("done")
```

```
got.wheelleg_start_balancing()
# move forward
got.wheelleg_move_speed_times(0, 30, 30, 1)
# move backward
got.wheelleg_move_speed_times(1, 30, 30, 1)
# turn left
got.wheelleg_turn_speed_times(2, 50, 90, 2)
# turn right
got.wheelleg_turn_speed_times(3, 50, 90, 2)
```

```
# drive in a square using the above commands
got.wheelleg_move_speed_times(0, 30, 30, 1)
got.wheelleg_turn_speed_times(2, 50, 90, 2)
got.wheelleg_move_speed_times(0, 30, 30, 1)
got.wheelleg_turn_speed_times(2, 50, 90, 2)
got.wheelleg_move_speed_times(0, 30, 30, 1)
got.wheelleg_turn_speed_times(2, 50, 90, 2)
got.wheelleg_move_speed_times(0, 30, 30, 1)
got.wheelleg_turn_speed_times(2, 50, 90, 2)
```

In programming, a key principle is DRY: Don't Repeat Yourself. The less you repeat yourself, the less likely you are to make mistakes and the easier it is to modify your code and understand it.

For loops are a way to repeat your code when you know how many times you should repeat it. (Later on we will learn about `while` loops, which should be used when we don't know how many times to repeat.)

As usual, in Python, you should pay attention to the colon and the indentation.

```
# for loop
for i in range(10):
    print(i)
    print("hi")
```

```
# shorter way of driving a square
for i in range(4):
    got.wheelleg_move_speed_times(0, 30, 30, 1)
    got.wheelleg_turn_speed_times(2, 50, 90, 2)
```

```
# drive an equilateral triangle
for i in range(3):
    got.wheelleg_move_speed_times(0, 30, 30, 1)
    got.wheelleg_turn_speed_times(2, 50, 120, 2)
```

You can see a pattern in the previous cells. We can use a variable to make this code more general:

```
# drive in a regular polygon
sides = 3
for i in range(sides):
    got.wheelleg_move_speed_times(0, 30, 30, 1)
    got.wheelleg_turn_speed_times(2, 50, int(360/sides), 2)
```
