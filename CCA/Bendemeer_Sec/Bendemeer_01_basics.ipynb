```
# this is a comment
# print("Bendemeer")
print(42) # this is another comment
```

```
# data types
print(type("hi")) # string
print(type(42)) # integer
print(type(4.2)) # float
print(type(True)) # boolean
print(type(False)) # boolean
```

```
# variables
name = "Michelle"
age = 80
print(f"I am {name}, {age} years old.")
print(f"In 5 years I'll be {age + 5}")

# exercise: store your hobby in a variable
hobby = "reading"
print(f"I like {hobby}")
```

```
# use a package
import random
roll = random.randint(1, 6)
print(roll)
```

```
# conditional statements
if False:
    print("the condition is true")
print("always printed")
```

```
# comparison operators
print(8 < 10) # less than
print(8 > 10) # more than
print(8 == 10) # equals
print(8 != 10) # not equals
```

```
age = 12
if age >= 13:
    print("You can watch Project Hail Mary")
else:
    print("you cannot watch")
```

```
# write a program to say if you pass/fail
grade = 75
if grade >= 50:
    print("pass")
else:
    print("fail")
```

```
# arithmetic operators
print(8 + 2)
print(8 - 2)
print(8 * 2)
print(8 / 2)

print("8" + "hi")
# print(8 + "9") # TypeError
print(8 * "hi")
```

```
# convert between types
print(str(8) + "9")
print(8 + int("9"))
```
