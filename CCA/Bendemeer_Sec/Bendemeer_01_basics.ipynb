## Comments

Comments are notes you write in your code to explain what it does. They start with the `#` symbol and Python ignores them when running the program. Comments help you remember what your code does and make it easier for others to understand.

```python
# this is a comment
# print("Bendemeer")
print(42) # this is another comment
```

## Types
Python categorises data into different **types**.
Some of the most common types are shown below.

```python
# data types
print(type("hi")) # string
print(type(42)) # integer
print(type(4.2)) # float
print(type(True)) # boolean
print(type(False)) # boolean
```

Note that Python is **case-sensitive**, i.e., capital / small letters matter.
For example, the following cell shows a common error:

```python
print(type(true))
```

## Variables
**Variables** are like labelled boxes that you can put things in.
Using variables makes your code easier to read, easier to modify, and less error-prone.
Below is an example of assigning variables (`name`, `age`, `hobby`) and using them in **f-strings**.

```python
# variables
name = "Michelle"
age = 80
print(f"I am {name}, {age} years old.")
print(f"In 5 years I'll be {age + 5}")

# exercise: store your hobby in a variable
hobby = "reading"
print(f"I like {hobby}")
```

## Packages
Sometimes, you will need functions that are not included in base Python.
Many **packages** have been written by other programmers in the past and made available for anyone to use.
To use a package, you usually just need to write `import` followed by the package name:

```python
# use a package
import random
roll = random.randint(1, 6)
print(roll)
```

## Conditional Statements

Conditional statements let your program make decisions. The most common one is `if`. It checks if something is true, and if it is, it runs the code inside. If not, it skips it. Think of it like: "If it's raining, take an umbrella." The code after the `if` only runs if the condition is true.

```python
# conditional statements
if False: # if you change this from False to True, it will print.
    print("the condition is true")
print("always printed")
```

## Comparison Operators

Comparison operators compare two values and return True or False. They're like asking questions: Is this bigger than that? Are they equal? These are essential for making decisions in your code.

```python
# comparison operators
print(8 < 10) # less than
print(8 > 10) # more than
print(8 == 10) # equals
print(8 != 10) # not equals
```

## Using Conditionals with Comparisons

Now let's combine comparison operators with if statements. This is how you make smart programs that react differently based on data.

```python
# watch a PG movie if you are over 13 years old
age = 12
if age >= 13:
    print("You can watch Project Hail Mary")
else:
    print("you cannot watch")
```

Let's write a program that decides if you pass or fail based on your grade. This shows how conditionals work in real situations.

```python
# write a program to say if you pass/fail
grade = 75
if grade >= 50:
    print("pass")
else:
    print("fail")
```

## Arithmetic Operators
You can do math in Python with numbers as usual.
What is maybe more surprising is that you can sometimes use the same operators on strings!

```python
# arithmetic operators
print(8 + 2)
print(8 - 2)
print(8 * 2)
print(8 / 2)

print("8" + "hi")
# print(8 + "9") # TypeError
print(8 * "hi")
```

## Type Conversion

Sometimes you need to change data from one type to another. For example, turning a number into text or text into a number. This is called type conversion and uses functions like `str()`, `int()`, and `float()`.

```python
# convert between types
print(str(8) + "9")
print(8 + int("9"))
```
