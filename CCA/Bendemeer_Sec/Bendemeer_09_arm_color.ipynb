This session's code combines arm movement, distance sensor usage and the AI color recognition module to pick up and throw a ball.

```python
from ugot import ugot
got = ugot.UGOT()
got.initialize("10.65.48.25") # 192.168.88.1

import time
got.load_models(["color_recognition"])

def pickup(): # define a function
    got.mechanical_clamp_release()
    time.sleep(1)
    # code the arm to pick up the ball from the table
    got.mechanical_joint_control(0, -40, -50, 500) # down
    time.sleep(1)
    got.mechanical_clamp_close()
    time.sleep(1)
    got.mechanical_joint_control(0, 30, 30, 500)  # up
    time.sleep(1)

def centralise():
    while True:
        info = got.get_color_total_info()
        if info:
            x_coord = info[2]
            if x_coord < 320: # ball is too far left
                got.mecanum_move_xyz(-5,0,0) # shift left
            elif x_coord > 320: # ball is too far right
                got.mecanum_move_xyz(5,0,0) # shift right
            else: # ball is exactly in the centre
                got.mecanum_stop()
                break # exit the while loop
def approach():
    IR = got.read_distance_data()
    while IR > 30:
        got.mecanum_move_speed(0, 5) # move forward slowly
        IR = got.read_distance_data()
        print(IR)
    got.mecanum_stop()

def throw():
    got.mechanical_joint_control(0, 80, 60, 500)
    time.sleep(1)
    got.mechanical_joint_control(0, 30, 30, 100)
    got.mechanical_clamp_release() 

centralise()
approach()
pickup() # call the function
throw()
```
