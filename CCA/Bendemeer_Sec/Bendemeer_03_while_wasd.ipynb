- `while` loop

The structure of the `while` loop is as follows:
```py
while condition:
    do something
```
The condition is something that evaluates to a boolean (`True` or `False`).
If the condition is `True`, the loop will keep executing the lines of code inside the loop.
Otherwise, the loop ends.

````python
i = 0 # create variable
while i < 10: # while (condition)
    print(i)
    i += 1 # i = i + 1
````

The previous cell functions just like a `for` loop, but you have to type more.
So what's the point of a `while` loop?

A: we use a `while` loop when we don't know how many times the loop should run.
The next cell shows an example:

````python
# we don't know how many tries it will take to enter the correct password
pw = ""
while pw != "password123":
    pw = input("Enter password")
print("welcome!")
````

Of course, rather than a condition, we can directly use `True` or `False` in the loop:

````python
# a special case of while loop
while True: # runs forever
    print("hi")
````

````python
while False: # will never run
    print("hi")
````

How do we get out of the loop without stopping it manually?
We use a special Python keyword called `break` (on its own line).
When Python sees `break`, it will immediately exit the innermost loop it is in.

````python
while True:
    number = input("enter number")
    if number == "42":
        print("correct")
        break # exit loop
        print("this will never run") # never prints since it is after break
````
