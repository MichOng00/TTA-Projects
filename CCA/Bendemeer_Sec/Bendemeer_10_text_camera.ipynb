```
import time
import cv2
import numpy as np
from ugot import ugot
got = ugot.UGOT()
got.initialize("10.65.48.47") # 192.168.88.1
got.open_camera()
got.load_models(["word_recognition"])
secret_word = "robot"
while True:
    frame = got.read_camera_data()
    nparr = np.frombuffer(frame, np.uint8) # NOT np.unit8
    data = cv2.imdecode(nparr, cv2.IMREAD_COLOR)
    guess = got.get_words_result()
    cv2.putText(data, f"{guess}", (20,20), cv2.FONT_HERSHEY_SIMPLEX,
                1, (255, 255, 255), 3, cv2.LINE_AA)
    if guess.isalpha() and len(guess) == len(secret_word):
        result = ""
        guess = guess.lower()
        for i in range(len(secret_word)):
            if guess[i] == secret_word[i]:
                result += "🟩"
            elif guess[i] in secret_word:
                result += "🟨"
            else:
                result += "◼️"
        print(result)
        print(guess)
        time.sleep(5) # import time at the top of the cell

        if guess == secret_word:
            print("you guessed it!")
            break
    cv2.imshow("UGOT camera", data)
    if cv2.waitKey(1) & 0xFF == ord('q'):  # if key q is pressed 
        break # exit window
cv2.destroyAllWindows()
```
