```
import pygame
pygame.init()
screen = pygame.display.set_mode((640,480)) # (width, height)
pygame.display.set_caption("UGOT control")

from ugot import ugot
got = ugot.UGOT()
got.initialize("192.168.88.1")
speed = 30
turn_speed = 45

running = True
while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False
    keys = pygame.key.get_pressed()
    y = 0
    if keys[pygame.K_w]:
        y = speed
    if keys[pygame.K_s]:
        y = -speed
    x = 0
    if keys[pygame.K_d]:
        x = speed
    if keys[pygame.K_a]:
        x = -speed
    z = 0
    if keys[pygame.K_q]:
        z = turn_speed
    if keys[pygame.K_e]:
        z = -turn_speed
    got.mecanum_move_xyz(x, y, z)

pygame.quit()

```
