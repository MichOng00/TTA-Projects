```
import random
import time
import pygame
pygame.init()
screen = pygame.display.set_mode((800, 600))
pygame.display.set_caption("Game window")
font = pygame.font.SysFont(None, 30)
hits = 0
start_time = time.time()
duration = 60
def get_time_remaining():
    elapsed_time = time.time() - start_time
    return max(0, int(duration - elapsed_time))
x, y = 400, 300
target_x, target_y = random.randint(0, 800), random.randint(0, 600)
speed = 1
running = True
while running:
    # Make sure you can quit
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False
    time_remaining = get_time_remaining()
    if time_remaining == 0:
        print(f"Your final score is {hits}")
        running = False
    keys = pygame.key.get_pressed()
    if keys[pygame.K_a]:
        x -= speed
    if keys[pygame.K_d]:
        x += speed
    if keys[pygame.K_w]:
        y -= speed
    if keys[pygame.K_s]:
        y += speed
    # limit the shape to the window
    x = max(0, min(720, x)) 
    y = max(0, min(520, y))

    screen.fill((247, 178, 2)) # RGB - background color
    target_rect = pygame.Rect(target_x, target_y, 40, 40)
    pygame.draw.rect(screen, (255,0,0), target_rect)
    player_rect = pygame.Rect(x, y, 80, 80)
    pygame.draw.rect(screen, (0,0,255), player_rect)

    # check collision
    if player_rect.colliderect(target_rect):
        hits += 1
        target_x, target_y = random.randint(0,760),random.randint(0,560)

    # pygame.draw.circle(screen, (0,0,255), (x, y), 30)
    hits_text = font.render(f"Hits: {hits}", True, (0,0,0))
    screen.blit(hits_text, (10, 10))
    timer_text = font.render(f"Time left: {time_remaining}",True,(0,0,0))
    screen.blit(timer_text, (10, 30))
    pygame.display.flip()

pygame.quit()
```
