```
from ugot import ugot
got = ugot.UGOT()
got.initialize("192.168.88.1")
# enable gesture recognition model
got.load_models(["gesture"])

# rock paper scissors
import random
# list to store options
options = ["Stone", "Scissors", "Cloth"]

# infinite game loop
while True:
    player = got.get_gesture_result()
    computer = random.choice(options)
    # make sure move is valid
    if player not in options:
        continue # skip the rest of the loop
    
    # possibility 1: player wins
    if (player == "Stone" and computer == "Scissors") or\
    (player == "Scissors" and computer == "Cloth") or\
    (player == "Cloth" and computer == "Stone"):
        print(f"you win! computer played {computer}")
        got.play_audio_tts("you win!", wait = True)
        got.screen_display_background(6) #green
    # possibility 2: draw
    elif player == computer:
        print(f"we tied! computer played {computer}")
        got.play_audio_tts("we tied!", wait = True)
        got.screen_display_background(5) #yellow
    # possibility 3: player loses
    else:
        print(f"you lose! computer played {computer}")
        got.play_audio_tts("you lose!", wait = True)
        got.screen_display_background(3) #red
    got.screen_display_background(0) #black
```

```
import numpy as np
import cv2
from ugot import ugot
got = ugot.UGOT()
got.initialize("10.66.154.184") # IP add
import time
got.open_camera()
got.load_models(["word_recognition"])
secret_word = "ROBOT"
attempts = 0
max_attempts = 6
while True:
    frame = got.read_camera_data()
    nparr = np.frombuffer(frame, np.uint8)
    data = cv2.imdecode(nparr, cv2.IMREAD_COLOR)
    text = got.get_words_result()
    cv2.putText(data, f"{text}", (20, 30), 0, 1,
                (0, 255, 0), 2, cv2.LINE_AA)
    # checks that the guess is the corect length
    # and everything in the text is a letter
    if len(text) == len(secret_word) and text.isalpha():
        attempts += 1
        result = ""
        text = text.upper() # make everything uppercase
        for i in range(len(secret_word)):
            # 1: correct letter in correct position
            if text[i] == secret_word[i]:
                result += "🟩"
                got.screen_display_background(6)
                got.play_tone("C6", 3)
            # 2: correct letter in WRONG position
            elif text[i] in secret_word:
                result += "🟨"
                got.screen_display_background(5)
                got.play_tone("G5", 3)
            # 3: letter is not in the word
            else:
                result += "⬛"
                got.screen_display_background(3) # red
                got.play_tone("C5", 3)
            # pause betweenn each letter
            # time.sleep(1)
            got.screen_display_background(0) # black
            # time.sleep(0.5)
        print(text, "\n", result)

        # exit program if you win
        if result == "🟩"*len(secret_word):
            print("You guessed it!")
            break
        # exit program if you are out of attempts
        if attempts >= max_attempts:
            print("Sorry, you are out of attempts")
            break
    cv2.imshow("camera", data)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
cv2.destroyAllWindows()
```
