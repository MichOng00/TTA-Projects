```
import random
print(random.randint(0, 5)) # random number from 0-5

# simulate rolling a dice
roll = random.randint(1, 6)
print(f"You rolled a {roll}")
```

```
# conditional statements
if True:
    print("the condition is true")

if False:
    print("the condition is false")

print(8 < 10) # True: 8 is less than 10
print(8 > 10) # False: 8 is more than 10
print(8 == 10) # False: 8 is not equal 10
print(8 != 10) # True: 8 is not equal to 10

var = False
if var:
    print("this is true")
else:
    print("this is false")
```

```
# create a variable called age
# ask for the user's age using input()
# convert it to a number using int()
age = int(input("What is your age? "))
# check if the user is above 13 years old
# if they are old enough, they can watch the movie.
# otherwise, they can't
if age > 13:
    print("You can watch thte movie!")
else: # age is less than or equal to 13
    print("Sorry, you can't watch the movie")
```
