```
my_variable = [2, "hello", True] # list

# print every item in the list
for thing in my_variable:
    print(thing)

# check if a specific item is in the list
if "hi" in my_variable:
    print("It's in the list")
else:
    print("It's not in the list")
```

```
from ugot import ugot
got = ugot.UGOT()
got.initialize("192.10.1.267") # change your IP address

got.load_models(["gesture"])
while True:
    print(got.get_gesture_result(), end="          \r")
```

```
import random

# list to store options
options = ["Stone", "Scissors", "Cloth"]

# infinite loop
while True:
    gesture = got.get_gesture_result()

    computer_choice = random.choice(options) # select a single item

    # possibility 1: player wins
    if gesture == "Stone" and computer_choice == "Scissors" or \
        gesture == "Scissors" and computer_choice == "Cloth" or \
        gesture == "Cloth" and computer_choice == "Stone":
        print(f"You win! I played {computer_choice}")
    # possibiity 2: draw
    elif gesture == computer_choice:
        print(f"We tied! I played {computer_choice}")
    # possibility 3: player loses
    else:
        print(f"You lose! I played {computer_choice}")
```
