```
from ugot import ugot
got = ugot.UGOT()
got.initialize("10.114.198.170")
```

```

got.show_light_rgb_effect(255, 0, 0, 3) # red, blue, green, effect
```

```
got.balance_start_balancing()
# move forward
got.balance_move_speed_times(0, 30, 30, 1)
# move backward
got.balance_move_speed_times(1, 30, 30, 1)
# turn left
got.balance_turn_speed_times(2, 50, 90, 2)
# turn right
got.balance_turn_speed_times(3, 50, 90, 2)
```

```
# drive in a square using the above commands
got.balance_move_speed_times(0, 30, 30, 1)
got.balance_turn_speed_times(2, 50, 90, 2)
got.balance_move_speed_times(0, 30, 30, 1)
got.balance_turn_speed_times(2, 50, 90, 2)
got.balance_move_speed_times(0, 30, 30, 1)
got.balance_turn_speed_times(2, 50, 90, 2)
got.balance_move_speed_times(0, 30, 30, 1)
got.balance_turn_speed_times(2, 50, 90, 2)
```

```
# drive in a square
for i in range(4):
    got.balance_move_speed_times(0, 30, 30, 1) # forward
    got.balance_turn_speed_times(2, 50, 90, 2) # left

# drive an equilateral triangle
for i in range(3): # change the number of times the loop runs
    got.balance_move_speed_times(0, 30, 30, 1) 
    got.balance_turn_speed_times(2, 50, 120, 2) # change the angle
```

```
# drive in a regular polygon
sides = 3
for i in range(sides): # change the number of times the loop runs
    got.balance_move_speed_times(0, 30, 30, 1) 
    got.balance_turn_speed_times(2, 50, int(360/sides), 2) # change the angle
```
