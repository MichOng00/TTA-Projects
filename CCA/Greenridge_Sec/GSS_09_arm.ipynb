```python
import time
from ugot import ugot
got = ugot.UGOT()
got.initialize("10.198.80.3")

got.mechanical_arms_restory()
```

```python
# gripper open / close
got.mechanical_clamp_release()
time.sleep(1)
got.mechanical_clamp_close()
```

```python
# figure out what positive / negative for each joint does
got.mechanical_joint_control(0, 0, 0, 500)
time.sleep(1)
got.mechanical_joint_control(30, 0, 0, 500) # joint 1 - positive left, negative right
time.sleep(1)
got.mechanical_joint_control(0, 30, 0, 500) # joint 2 - positive up, negative down
time.sleep(1)
got.mechanical_joint_control(0, 0, 30, 500) # joint 3 - positive up, negative down
time.sleep(1)
```
