```
# while loop examples
i = 0
while i < 10: # condition is either True or False
    print(i)
    i += 1
```

When to use `for` vs `while`?

Use `while` when you don't know how many times the loop will run.

```
# example: password
pw = ""
while pw != "password123":
    pw = input("Please enter the password: ")
```

```
# a special kind of while loop
i = 0
while True: # runs forever
    print(i)
    i += 1
    if i == 10:
        break # exits the loop
```

```
from ugot import ugot
got = ugot.UGOT()
got.initialize("192.168.88.1")
```

```
while True:
    distance = got.read_distance_data(51)
    print(f"{distance}", end = "     \r")
```

```
import time
got.balance_start_balancing()
time.sleep(2)
distance = got.read_distance_data(51)

# move forward until obstacle in front
while distance > 20:
    distance = got.read_distance_data(51)
    got.balance_move_speed(0, 20)

got.balance_stop_balancing()
```

```
got.balance_start_balancing()
time.sleep(2)
distance = got.read_distance_data(51)

# keep obstacle at constant distance
while True:
    distance = got.read_distance_data(51)
    if distance > 45:
        got.balance_move_speed(0, 20)
    elif distance < 35:
        got.balance_move_speed(1, 20)
    else:
        got.balance_stop_balancing()

got.balance_stop_balancing()
```
