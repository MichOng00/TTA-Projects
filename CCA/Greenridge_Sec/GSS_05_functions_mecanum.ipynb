```
# functions with parameters
def function_with_parameters(param1, param2):
    print(param1, param2)

def greet(name):
    print(f"hello, {name}")

greet("GSS")
```

```
# example: multiply by five
def multiply_five(num):
    result = num * 5
    print(f"the result is {result}")

multiply_five(3)

# example: add two numbers
def add_numbers(num1, num2):
    print(f"the result is {num1 + num2}")

add_numbers(5, 9)
```

```
from ugot import ugot
got = ugot.UGOT()
got.initialize("192.168.88.1")
```

```
import time

got.mecanum_move_xyz(30, 0, 0) # positive x
time.sleep(3)
got.mecanum_move_xyz(-30, 0, 0) # negative x
time.sleep(3)
got.mecanum_move_xyz(0, 30, 0) # positive y
time.sleep(3)
got.mecanum_move_xyz(0, -30, 0) # negative y
time.sleep(3)
got.mecanum_move_xyz(0, 0, 30) # positive z
time.sleep(3)
got.mecanum_move_xyz(0, 0, -30) # negative z
time.sleep(3)
got.mecanum_stop() # don't forget the stop!!
```

```
# combining two parameters
got.mecanum_move_xyz(30, 30, 0) # try different combinations
time.sleep(3)
got.mecanum_stop()

# challenge: drive the robot in 3 different types of circles
# 1) camera pointing outwards
# 2) camera pointing towards the centre of the circle
# 3) camera facing direction of travel (along the circle)
def camera_outwards():
    got.mecanum_move_xyz(20, 0, -60) # try different combinations
    time.sleep(3)
    got.mecanum_stop()
def camera_inwards():
    got.mecanum_move_xyz(-20, 0, -60) # try different combinations
    time.sleep(3)
    got.mecanum_stop()
def camera_forwards():
    got.mecanum_move_xyz(0, 20, 60) # try different combinations
    time.sleep(3)
    got.mecanum_stop()

camera_forwards()
```
