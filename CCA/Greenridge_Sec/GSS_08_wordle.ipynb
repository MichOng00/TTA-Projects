```python
# text recognition from last time
import numpy as np
import cv2

from ugot import ugot
got = ugot.UGOT()
got.initialize("192.168.88.1")

got.load_models(["word_recognition"])

got.open_camera()

while True:
    frame = got.read_camera_data()
    nparr = np.frombuffer(frame, np.uint8)
    data = cv2.imdecode(nparr, cv2.IMREAD_COLOR)

    guess = got.get_words_result()
    cv2.putText(data, f"{guess}", (20, 20), cv2.FONT_HERSHEY_SIMPLEX, 
                1, (255, 255, 255), 3, cv2.LINE_AA)

    cv2.imshow("UGOT camera", data)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cv2.destroyAllWindows()
```

```python
# wordle - attempts can be exercise?
import numpy as np
import cv2
import time

from ugot import ugot
got = ugot.UGOT()
got.initialize("192.168.88.1")

got.load_models(["word_recognition"])

got.open_camera()

secret_word = "ROBOT"
# attempts = 0
# max_attempts = 5

while True:
    frame = got.read_camera_data()
    nparr = np.frombuffer(frame, np.uint8)
    data = cv2.imdecode(nparr, cv2.IMREAD_COLOR)

    guess = got.get_words_result()
    cv2.putText(data, f"{guess}", (20, 20), cv2.FONT_HERSHEY_SIMPLEX, 
                1, (255, 255, 255), 3, cv2.LINE_AA)
    
    if guess.isalpha() and len(guess) == len(secret_word):
        # attempts += 1
        result = ""
        guess = guess.upper()
        for i in range(len(secret_word)):
            if guess[i] == secret_word[i]:
                result += "🟩"
                got.screen_display_background(6) # green
                got.play_tone("C6", 3) # high C
            elif guess[i] in secret_word:
                result += "🟨"
                got.screen_display_background(5) # yellow
                got.play_tone("G5", 3)
            else:
                result += "🟥"
                got.play_tone("C5", 3)
                got.screen_display_background(3) # red

            # pause in between colors
            time.sleep(1)
            got.screen_display_background(0) # black
            time.sleep(1)

        print(guess, "\n", result)
        time.sleep(1)

        # exit conditions
        if result == "🟩🟩🟩🟩🟩":
            print("You guessed it!")
            # print(f"You took {attempts} attempts.")
            break
        # elif attempts >= max_attempts:
        #     print("Sorry, you are out of attempts.")
        #     break

    cv2.imshow("UGOT camera", data)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cv2.destroyAllWindows()
```
