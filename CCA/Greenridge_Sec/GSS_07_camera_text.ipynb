```python
# navigation
import numpy as np
import cv2

from ugot import ugot
got = ugot.UGOT()
got.initialize("192.168.88.1")

got.load_models(["word_recognition"])

got.open_camera()

while True:
    frame = got.read_camera_data()
    nparr = np.frombuffer(frame, np.uint8)
    data = cv2.imdecode(nparr, cv2.IMREAD_COLOR)

    text = got.get_words_result()
    cv2.putText(data, f"{text}", (20, 20), cv2.FONT_HERSHEY_SIMPLEX, 
                1, (255, 255, 255), 3, cv2.LINE_AA)
    
    # check directions - can do == vs in versions
    if "left" in text:
        got.mecanum_turn_speed(2, 30)
    elif "right" in text:
        got.mecanum_turn_speed(3, 30)
    # etc - can also include other directions, light up

    cv2.imshow("UGOT camera", data)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cv2.destroyAllWindows()
```
