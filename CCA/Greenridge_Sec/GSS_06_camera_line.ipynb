```python
# using the camera
import numpy as np
import cv2

from ugot import ugot
got = ugot.UGOT()
got.initialize("192.168.88.1")

got.open_camera()
# got.load_models(["line_recognition"])
# got.set_track_recognition_line(0)

while True:
    frame = got.read_camera_data()
    nparr = np.frombuffer(frame, np.uint8)
    data = cv2.imdecode(nparr, cv2.IMREAD_COLOR)

    # 1) show offset
    line_info = got.get_single_track_total_info()
    if line_info:
        offset = line_info[0]
        cv2.putText(data, f"Offset: {offset}", (50, 40), 
                cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)
    
        # 2) follow line - let them go through errors
        rotation_speed = int(offset * 0.5)
        got.mecanum_move_xyz(0, 20, rotation_speed)
        # 3) stop at no line
        line_type = line_info[1]
        if line_type == 0:
            got.mecanum_stop()
    

    cv2.imshow("UGOT camera", data)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cv2.destroyAllWindows()
```
