```python
from ugot import ugot
got = ugot.UGOT()
got.initialize("192.168.1.xxx")

import time
```

```python
got.mechanical_arms_restory() # reset the arm
```

```python
got.mechanical_clamp_release() # open the gripper
time.sleep(1)                  # wait 1 second
got.mechanical_clamp_close()   # close the gripper
```

```python
got.mechanical_joint_control(angle=0, angle2=0, angle3=0, duration=500)
```

```python
def pick_up():
    got.mechanical_clamp_release()
    time.sleep(1)
    got.mechanical_joint_control(0, -60, 0, 500) # change angles to go down
    time.sleep(1)
    got.mechanical_clamp_close()
    time.sleep(1)
    got.mechanical_joint_control(0, 0, 0, 500) # change angles to go up
pick_up()
```

```python
def throw():
    got.mechanical_joint_control(0, 40, 40, 50)
    time.sleep(1)
    got.mechanical_joint_control(0, 0, 0, 50)
    got.mechanical_clamp_release()

throw()
```

```python
import random
words = ["world", "robot", "light", "bytes"]
word = random.choice(words)
guessed = ["_" for letter in word] # list comprehension - combine for loop with a list
print(guessed)
attempts = 6

while attempts > 0 and "_" in guessed:
    print(f"Word: {" ".join(guessed)}")
    guess = input("Guess a letter: ")
    if guess in word:
        for i in range(len(word)):
            if word[i] == guess: # if the letter in the actual word is equal to the guessed letter
                guessed[i] = guess 
    else:
        attempts -= 1
        print(f"Wrong, you have {attempts} attempts left")

# OPTIONAL
if "_" not in guessed:
    print(f"Congrats! You found the word: {word}")
else:
    print(f"Game over! The word was: {word}")
```

```python
import random
import time
options = ["Stone", "Scissors", "Cloth"]
while True:
    gesture = got.get_gesture_result()
    player_choice = gesture
    computer_choice = random.choice(options)

    if player_choice in options: # check that player choice is valid
        if (player_choice == "Cloth" and computer_choice == "Stone") or\
        (player_choice == "Stone" and computer_choice == "Scissors") or\
        (player_choice == "Scissors" and computer_choice == "Cloth"):
            print(f"You win! I played {computer_choice}")
        # EXERCISE: finish the code (tie and lose)
        elif player_choice == computer_choice:
            print(f"We draw! I played {computer_choice}")
        else:
            print(f"You lose! I played {computer_choice}")

        time.sleep(2)
        
```

```python
from ugot import ugot
got = ugot.UGOT()
got.initialize("192.168.1.130")

got.load_models(["gesture"])
```

```python
options = ["Cloth", "Stone", "Scissors", "OK", "Likes"]
while True:
    gesture = got.get_gesture_result()
    if gesture == "Stone":
        got.mecanum_move_speed(0, 15) # move forward
    # exercise: code the other gestures (Scissors - backward, Cloth - stop, OK - turn left, Likes - turn right)
    elif gesture == "Scissors":
        got.mecanum_move_speed(1, 15) # move backward
    elif gesture == "OK":
        got.mecanum_turn_speed(2, 45) # turn left
    elif gesture == "Likes":
        got.mecanum_turn_speed(3, 45) # turn right
    elif gesture == "Cloth":
        got.mecanum_stop()
    else: # if the robot doesn't recognise any gesture
        got.mecanum_stop()
```

```python
options = ["Cloth", "Stone", "Scissors", "OK", "Likes"]
while True:
    go = False
    commands = []
    while not go:
        gesture = got.get_gesture_result()
        if gesture in options[1:]: # everything except Cloth
            commands.append(gesture)
            got.play_audio_tts(f"Added {gesture}", wait=True)
            print(commands)
        elif gesture == "Cloth":
            got.play_audio_tts("Commands activated", wait=True)
            go = True

    for c in commands:
        if c == "Stone":
            got.mecanum_move_speed_times(0, 15, 30, 1) # move forward 30 cm
        elif c == "Scissors":
            got.mecanum_move_speed_times(1, 15, 30, 1) # move backward 30 cm
        elif c == "OK":
            got.mecanum_turn_speed_times(2, 30, 90, 2) # turn left 90 degrees
        else:
            got.mecanum_turn_speed_times(3, 30, 90, 2) # turn right 90 degrees
```
