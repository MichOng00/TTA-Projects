```python
import pygame

pygame.init()

screen = pygame.display.set_mode((800, 600))
pygame.display.set_caption("hello")

x, y = 100, 100
speed = 1

running = True
while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    keys = pygame.key.get_pressed()
    # why not if-elif? cannot move diagonally
    if keys[pygame.K_LEFT]: # pygame.K_a
        x -= speed
    if keys[pygame.K_RIGHT]: # pygame.K_d
        x += speed
    if keys[pygame.K_UP]: # pygame.K_w
        y -= speed
    if keys[pygame.K_DOWN]: # pygame.K_s
        y += speed

    # wrap around edges
    if x < 0:
        x = 800
    elif x > 800:
        x = 0
    if y < 0:
        y = 600
    elif y > 600:
        y = 0
    

    screen.fill((255, 255, 255))

    # draw shapes
    # rect (surface, color, position (x, y, width, height))
    pygame.draw.rect(screen, (0, 0, 255, 50), (100, 100, 200, 100))
    # circle (surface, color, centre (x, y), radius)
    pygame.draw.circle(screen, (255, 0, 0), (x, y), 50)
    # polygon (surface, color, points)
    pygame.draw.polygon(screen, (255, 100, 0), [(100, 150), (200, 150), (150, 300)])


    # update display
    pygame.display.flip()

pygame.quit()
```
