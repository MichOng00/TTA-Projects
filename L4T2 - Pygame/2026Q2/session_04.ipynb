```python
import random
import pygame
pygame.init()
pygame.mixer.init() # for sound

# pygame.mixer.music.load("filename.mp3") # background music
# pygame.mixer.music.play(-1) # loop

collision_sound = pygame.mixer.Sound("collide.mp3")

game_duration = 30 * 1000 # 30 seconds in ms
start_time = pygame.time.get_ticks()

screen = pygame.display.set_mode((800, 600))
pygame.display.set_caption("Dodge the enemy")

player_img = pygame.image.load("zelda.png")
player_img = pygame.transform.scale(player_img, (50, 50))
enemy_img = pygame.image.load("ganon.png")
enemy_img = pygame.transform.scale(enemy_img, (100, 100))

player_pos = [100, 100]
enemy_pos = [300,300]
obstacle_pos = [150, 150]

font = pygame.font.SysFont("Arial", 30)

enemy_direction = [random.choice([-1, 1]), random.choice([-1, 1])]
enemy_speed = 2

hits = 0

speed = 3
clock = pygame.time.Clock()

running = True
game_over = False
while running:
    while game_over:
        pygame.mixer.stop()
        screen.fill((255, 255, 255))
        if hits < 3:
            result_text = font.render("You win! Press 'C' to play again.", True, (0, 255, 0))
        else:
            result_text = font.render("You lost! Press 'C' to play again.", True, (255, 0, 0))
        screen.blit(result_text, (screen.get_width() // 2 - result_text.get_width() // 2, 200))
        pygame.display.flip()

        for event in pygame.event.get():
            if event.type == pygame.QUIT:
                running = False
                game_over = False
            if event.type == pygame.KEYDOWN:
                if event.key == pygame.K_c:
                    game_over = False
                    player_pos = [100, 100]
                    enemy_pos = [random.randint(0, 800), random.randint(0, 600)]
                    hits = 0
                    start_time = pygame.time.get_ticks()

    elapsed_time = pygame.time.get_ticks() - start_time
    remaining_time = max(0, (game_duration - elapsed_time) // 1000)

    if remaining_time == 0 or hits >= 3:
        game_over = True

    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    # player movement
    keys = pygame.key.get_pressed()
    if keys[pygame.K_LEFT]: # K_a
        player_pos[0] -= speed
    if keys[pygame.K_RIGHT]:
        player_pos[0] += speed
    if keys[pygame.K_UP]:
        player_pos[1] -= speed
    if keys[pygame.K_DOWN]:
        player_pos[1] += speed

    # enemy movement
    enemy_pos[0] += enemy_direction[0] * enemy_speed
    enemy_pos[1] += enemy_direction[1] * enemy_speed

    # bounce the enemy off the window edges
    if enemy_pos[0] <= 0 or enemy_pos[0] + 100 >= 800:
        enemy_direction[0] *= -1
        # enemy_speed += 0.1
    if enemy_pos[1] <= 0 or enemy_pos[1] + 100 >= 600:
        enemy_direction[1] *= -1
        # enemy_speed += 0.1

    # limit player to the window
    if player_pos[0] < 0:
        player_pos[0] = 0
    elif player_pos[0] + 50 >= 800:
        player_pos[0] = 800 - 50
    if player_pos[1] < 0:
        player_pos[1] = 0
    elif player_pos[1] + 50 >= 600:
        player_pos[1] = 600 - 50

    screen.fill((255, 255, 255))
    # display images
    screen.blit(player_img, player_pos)
    screen.blit(enemy_img, enemy_pos)

    # draw rectangles
    player_rect = pygame.Rect(*player_pos, 50, 50) # unpack the list
    enemy_rect = pygame.Rect(*enemy_pos, 100, 100)
    pygame.draw.rect(screen, (0, 255, 0), player_rect, width = 3)
    pygame.draw.rect(screen, (0, 255, 0), enemy_rect, width = 3)

    # handle collision
    collision = player_rect.colliderect(enemy_rect)
    if collision:
        collision_sound.play()
        hits += 1
        screen.blit(font.render("You got hit!", True, (255, 0, 0)), player_pos)
        pygame.draw.rect(screen, (255, 0, 0), player_rect, width = 3)
        pygame.display.flip()
        pygame.time.wait(1000)
        # respawn
        player_pos = [random.randint(0, 800), random.randint(0, 600)]

    # obstacle collision (enemy only)
    obstacle_rect = pygame.Rect(*obstacle_pos, 100, 100)
    pygame.draw.rect(screen, (0, 255, 0), obstacle_rect, width = 3)
    obs_collision = obstacle_rect.colliderect(enemy_rect)
    if obs_collision:
        delta_x = obstacle_rect.centerx - enemy_rect.centerx
        delta_y = obstacle_rect.centery - enemy_rect.centery
        if abs(delta_x) > abs(delta_y):
            enemy_direction[0] *= -1
        else:
            enemy_direction[1] *= -1

        obstacle_pos = [random.randint(0, 800-100), random.randint(0, 600-100)]

    # display number of times hit
    hits_text = font.render(f"Hits: {hits}", True, (0, 0, 0))
    time_text = font.render(f"Time: {remaining_time}s", True, (0, 0, 0))
    screen.blit(hits_text, (10, 10))
    screen.blit(time_text, (10, 50))

    pygame.display.flip()
    clock.tick(60)

pygame.quit()
```
