```python
print("hello")
print(type("hello"))  # string - usually words / letters
print(42)
print(type(42))       # integer - whole number
print(type(42.0))     # float - decimal
print(type("42"))     # string
print(type(True))     # boolean
print(type(False))     # boolean
```

```python
# arithmetic operators
print(6 + 2)
print(6 - 2)
print(6 * 2) # multiply
print(6 / 2) # divide ALWAYS gives you a float
print(int(6 / 2)) # convert from a float to an integer
```

```python
# comparison operators - give a Boolean 
print(6 < 7) # less than
print(6 > 7) # more than
print(6 == 7) # equal
print(6 != 7) # not equal

print("hello" == "Hello") # case-sensitive (capital vs small matters!)
print(3 == 3.0)           # can compare integers and floats
```

```python
# conditional statements
if True: 
    print("this is true") # will not print if the condition is False

print("this is always printed")
```

```python
if False: 
    print("this is true")
else: # don't need a condition
    print("this is false")
```

```python
age = 10
if age > 13:
    print("You can watch spiderman 🕷️ :D")
else:
    print("You are too young :(") 
```

```python
# more than one condition
if False:
    print("1st condition is true")
elif True:
    print("1st condition is false AND 2nd condition is true")
else:
    print("1st condition is false AND 2nd condition is also false")
```

```python
# Exercise: write a program that prints a grade (A/B/fail) based on your mark
mark = input("Enter your mark")
if mark > 80:
    print("A")
elif mark > 60:
    print("B")
else:
    print("fail")
```

```python
mark = int(input("Enter your mark"))
if mark > 80:
    print("A")
elif mark > 60:
    print("B")
elif mark > 40:
    print("C")
else:
    print("fail")
```

```python
# logical operators
# and - both have to be True 
print(True and True)
print(True and False)
print(False and False)

# or - at least one has to be True
print(True or True)
print(True or False)
print(False or False)

# not - the opposite
print(not True)
print(not False)
```
