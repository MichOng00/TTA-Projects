```python
# number guessing with multiple guesses
# exercise: print out how many attempts you took in the end
import random
random_number = random.randint(1, 20)
times = 0
max_attempts = 5
while True:
    times += 1
    user_guess = int(input("Guess a number from 1 to 20"))
    print(f"Your guess: {user_guess}")
    if random_number == user_guess:
        print("Correct")
        print(f"You guessed it in {times} attempts")
        break
    elif user_guess > random_number:
        print("Too high")
    else: # don't need user_guess < random_number because that is the only other possibility
        print("Too low")

    if times > max_attempts:
        print("Sorry, you are out of attempts.")
        break
```

```python
# example: to-do list
todo = []

while True:
    command = input("Enter a command (add / show / done / quit):")
    if command == "add":
        task = input("Enter the new task:")
        todo.append({"task": task, "done": False})
    elif command == "quit":
        print("Goodbye!")
        break
    elif command == "show":
        if len(todo) == 0:
            print("Nothing to do!")
        else:
            for i, task in enumerate(todo):
                mark = "x" if task["done"] else " "
                print(f"{i+1}. [{mark}] {task["task"]}")
    elif command == "done":
        completed = int(input("Which number task is done?"))
        todo[completed - 1]["done"] = True
    else:
        print("Unknown command.")
```

```python
import pygame
pygame.init()
screen = pygame.display.set_mode((800, 600)) # (width, height)
pygame.display.set_caption("shapes") # window title

x = 600
y = 300
speed = 4
running = True
while running:
    # clear the event queue
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    keys = pygame.key.get_pressed()
    if keys[pygame.K_a]:
        x -= speed
    if keys[pygame.K_d]:
        x += speed
    # exercise: w to go up, s to go down
    if keys[pygame.K_w]:
        y -= speed
    if keys[pygame.K_s]:
        y += speed

    # limit the shape to the window
    if x < 80:
        x = 80
    if x > 800 - 80:
        x = 800 - 80
    # exercise: let the shape not go out the top and bottom
    if y < 80:
        y = 80
    if y > 600 - 80:
        y = 600 - 80

    screen.fill((255, 255, 255)) # (red, green, blue) 0-255

    # draw shapes
    pygame.draw.circle(screen, (12, 110, 80), (x, y), 80)
    pygame.draw.rect(screen, (0, 0, 255), (100, 100, 200, 100)) # (x, y, width, height)

    # update display
    pygame.display.flip()

# print(keys)
pygame.quit()
```

```python
import random
import pygame
pygame.init()
screen = pygame.display.set_mode((800, 600)) # (width, height)
pygame.display.set_caption("shapes") # window title

player_img = pygame.image.load("firefox.png")
player_img = pygame.transform.scale(player_img, (50, 50))
enemy_img = pygame.image.load("chrome.png")
can_img = pygame.transform.scale(can_img, (50, 50))

x = 600
y = 300
enemy_x = 100
y_y = 100
enemy_pos = [100, 100]
speed = 2
enemy_direction = [random.choice([-1, 1]), random.choice([-1, 1])]
enemy_speed = 2
running = True
clock = pygame.time.Clock()
while running:
    # clear the event queue
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    keys = pygame.key.get_pressed()
    if keys[pygame.K_a]:
        x -= speed
    if keys[pygame.K_d]:
        x += speed
    # exercise: w to go up, s to go down
    if keys[pygame.K_w]:
        y -= speed
    if keys[pygame.K_s]:
        y += speed

    # limit the shape to the window
    if x < 0:
        x = 0
    if x > 800 - 50:
        x = 800 - 50
    # exercise: let the shape not go out the top and bottom
    if y < 0:
        y = 0
    if y > 600 - 50:
        y = 600 - 50

    # enemy movement
    enemy_pos[0] += enemy_direction[0] * enemy_speed
    enemy_pos[1] += enemy_direction[1] * enemy_speed
    # bounce enemy off the window edges
    if enemy_pos[0] <= 0 or enemy_pos[0] + 50 >= 800:
        enemy_direction[0] *= -1
    if enemy_pos[1] <= 0 or enemy_pos[1] + 50 >= 600:
        enemy_direction[1] *= -1

    screen.fill((255, 255, 255)) # (red, green, blue) 0-255

    screen.blit(player_img, (x, y))
    screen.blit(can_img, enemy_pos)

    player_rect = pygame.Rect(x, y, 50, 50)
    enemy_rect = pygame.Rect(*enemy_pos, 50, 50)
    pygame.draw.rect(screen, (0, 255, 0), player_rect, width = 3)
    pygame.draw.rect(screen, (0, 255, 0), enemy_rect, width = 3)

    # change border color if they collide
    if player_rect.colliderect(enemy_rect):
        pygame.draw.rect(screen, (255, 0, 0), player_rect, width = 3)

    # update display
    pygame.display.flip()
    clock.tick(60)

# print(keys)
pygame.quit()
```

```python
import random
import pygame
pygame.init()
pygame.mixer.init()

collision_sound = pygame.mixer.Sound("collision.mp3")
bounce_sound = pygame.mixer.Sound("bounce.mp3")

game_duration = 30 * 1000
start_time = pygame.time.get_ticks() # current time
screen = pygame.display.set_mode((800, 600)) # (width, height)
pygame.display.set_caption("shapes") # window title

player_img = pygame.image.load("firefox.png")
player_img = pygame.transform.scale(player_img, (50, 50))
enemy_img = pygame.image.load("chrome.png")
enemy_img = pygame.transform.scale(enemy_img, (50, 50))

x = 600
y = 300
enemy_x = 100
enemy_y = 100
enemy_pos = [100, 100]
obstacle_pos = [random.randint(0, 800-100), random.randint(0, 600-100)]
font = pygame.font.SysFont(None, 30)
speed = 2
enemy_direction = [random.choice([-1, 1]), random.choice([-1, 1])]
enemy_speed = 2
hits = 0
running = True
game_over = False
clock = pygame.time.Clock()
while running:
    while game_over:
        screen.fill((255, 255, 255))
        if hits < 3:
            result_text = font.render("You win! Press 'C' to play again.", True, (0, 255, 0))
        else:
            result_text = font.render("You lose! Press 'C' to play again.", True, (255, 0, 0))
        screen.blit(result_text, (screen.get_width() // 2 - result_text.get_width() // 2, 200))
        pygame.display.flip()
        for event in pygame.event.get():
            if event.type == pygame.QUIT:
                running = False
                game_over = False
            if event.type == pygame.KEYDOWN:
                if event.key == pygame.K_c:
                    game_over = False
                    x = 100
                    y = 100
                    enemy_pos = [random.randint(0, 800), random.randint(0, 600)]
                    hits = 0
                    start_time = pygame.time.get_ticks()
    elapsed_time = pygame.time.get_ticks() - start_time
    remaining_time = max(0, (game_duration - elapsed_time) // 1000) # floor division
    if remaining_time == 0 or hits >= 3:
        game_over = True
    # clear the event queue
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    keys = pygame.key.get_pressed()
    if keys[pygame.K_a]:
        x -= speed
    if keys[pygame.K_d]:
        x += speed
    # exercise: w to go up, s to go down
    if keys[pygame.K_w]:
        y -= speed
    if keys[pygame.K_s]:
        y += speed

    # limit the shape to the window
    if x < 0:
        x = 0
    if x > 800 - 50:
        x = 800 - 50
    # exercise: let the shape not go out the top and bottom
    if y < 0:
        y = 0
    if y > 600 - 50:
        y = 600 - 50

    # enemy movement
    enemy_pos[0] += enemy_direction[0] * enemy_speed
    enemy_pos[1] += enemy_direction[1] * enemy_speed
    # bounce enemy off the window edges
    if enemy_pos[0] <= 0 or enemy_pos[0] + 50 >= 800:
        enemy_direction[0] *= -1
    if enemy_pos[1] <= 0 or enemy_pos[1] + 50 >= 600:
        enemy_direction[1] *= -1

    screen.fill((255, 255, 255)) # (red, green, blue) 0-255

    screen.blit(player_img, (x, y))
    screen.blit(enemy_img, enemy_pos)

    player_rect = pygame.Rect(x, y, 50, 50)
    enemy_rect = pygame.Rect(*enemy_pos, 50, 50)
    pygame.draw.rect(screen, (0, 255, 0), player_rect, width = 3)
    pygame.draw.rect(screen, (0, 255, 0), enemy_rect, width = 3)

    # change border color if they collide
    if player_rect.colliderect(enemy_rect):
        collision_sound.play()
        pygame.draw.rect(screen, (255, 0, 0), player_rect, width = 3)
        hits += 1
        screen.blit(font.render("You got hit!", True, (255, 0, 0)), (x, y))
        pygame.display.flip()
        pygame.time.wait(1000) # wait 1 second
        # respawn at a random position
        x = random.randint(0, 800)
        y = random.randint(0, 600)

    # safe zone
    obstacle_rect = pygame.Rect(*obstacle_pos, 100, 100)
    pygame.draw.rect(screen, (0, 255, 0), obstacle_rect, width=3)
    if obstacle_rect.colliderect(enemy_rect):
        bounce_sound.play()
        # make enemy bounce off the side
        delta_x = obstacle_rect.centerx - enemy_rect.centerx
        delta_y = obstacle_rect.centery - enemy_rect.centery
        if abs(delta_x) > abs(delta_y):
            enemy_direction[0] *= -1
        else:
            enemy_direction[1] *= -1
        obstacle_pos = [random.randint(0, 800-100), random.randint(0, 600-100)]

    hits_text = font.render(f"Hits: {hits}", True, (0,0,0))
    screen.blit(hits_text, (10, 10))
    time_text = font.render(f"Time: {remaining_time}", True, (0,0,0))
    screen.blit(time_text, (10, 50))
    # update display
    pygame.display.flip()
    clock.tick(60)

# print(keys)
pygame.quit()
```

```python
import pygame
import random
import time
import os

pygame.init()

screen = pygame.display.set_mode((800, 600))
pygame.display.set_caption("Catch game")

player_img = pygame.image.load("basket.png")
player_img = pygame.transform.scale(player_img, (100, 50))
player2_img = pygame.transform.scale(player_img, (100, 50))
apple_img_1 = pygame.image.load("1apple.png")
apple_img_1 = pygame.transform.scale(apple_img_1, (50, 50))
apple_img_5 = pygame.image.load("5apple.png")
apple_img_5 = pygame.transform.scale(apple_img_5, (50, 50))
can_img = pygame.image.load("can.png")
can_img = pygame.transform.scale(can_img, (50, 50))
mult_img = pygame.image.load("green_apple.jpg")
mult_img = pygame.transform.scale(mult_img, (50, 50))

player_speed = 7
player_x = 350
player_y = 500
player2_x = 400
player2_y = 500
score = 0
start_time = time.time()
duration = 60
high_score_file = "highscores.txt"
mult_active = False
mult_end_time = 0
multiplier = 2 

def get_time_remaining():
    elapsed_time = time.time() - start_time
    return max(0, int(duration - elapsed_time))

clock = pygame.time.Clock()

choices_list = []
choices_list.append(can_img)
choices_list.append(mult_img)
for _ in range(3):
    choices_list.append(apple_img_1)
    choices_list.append(apple_img_5)

apples = [
    {"img" : random.choice(choices_list),
     "x" : random.randint(0, 750), 
     "y" : random.randint(-600, -50),
     "speed" : random.randint(3, 7)}
     for _ in range(5)
]

apples_copy = apples.copy()

# high score helpers
def load_high_score():
    if os.path.exists(high_score_file):
        with open(high_score_file, "r") as file: # r for read
            return int(file.read().strip() or 0)
    return 0

def save_high_score(score):
    with open(high_score_file, "w") as file: # w for write
        file.write(str(score))

high_score = load_high_score()
font = pygame.font.Font(None, 36)

game_over = True
running = True
while running:
    while game_over:
        screen.fill((255, 255, 255))
        game_over_text = font.render("Press n for single player, m for multiplayer", True, (0,0,0))
        screen.blit(game_over_text, (screen.get_width()//2 - game_over_text.get_width()//2, 200))
        pygame.display.flip()
        for event in pygame.event.get():
            if event.type == pygame.QUIT:
                running = False
                game_over = False

            if event.type == pygame.KEYDOWN:
                if event.key == pygame.K_n: # no multiplayer
                    multiplayer = False
                    score = 0
                    player_x = 350
                    apples = apples_copy.copy()
                    mult_active = False
                    start_time = time.time()
                    game_over = False
                elif event.key == pygame.K_m: # multiplayer
                    multiplayer = True
                    score = 0
                    player_x = 350
                    apples = apples_copy.copy()
                    mult_active = False
                    start_time = time.time()
                    game_over = False


    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    keys = pygame.key.get_pressed()
    if keys[pygame.K_LEFT] and player_x > 0:
        player_x -= player_speed
    if keys[pygame.K_RIGHT] and player_x < 800 - 100:
        player_x += player_speed

    if multiplayer:
        if keys[pygame.K_a] and player2_x > 0:
            player2_x -= player_speed
        if keys[pygame.K_d] and player2_x < 800 - 100:
            player2_x += player_speed

    time_remaining = get_time_remaining()
    if mult_active and time.time() >= mult_end_time:
        mult_active = False
        multiplier = 2
        player_speed = 7

    if time_remaining == 0:
        print(f"Your final score is {score}")
        if score > high_score:
            high_score = score
            save_high_score(high_score)
            print("New high score!")
        game_over = True

    screen.fill((255, 255, 255))

    player_rect = pygame.Rect(player_x, player_y, 100, 50)
    if multiplayer:
        player2_rect = pygame.Rect(player2_x, player2_y, 100, 50)
    else:
        player2_rect = pygame.Rect(player_x, player_y, 100, 50)
    # handle apples
    for apple in apples:
        apple["y"] += apple["speed"]
        apple_rect = pygame.Rect(apple["x"], apple["y"], 50, 50)
        collision = player_rect.colliderect(apple_rect)
        collision2 = player2_rect.colliderect(apple_rect)
        if apple["y"] > 600 or collision or collision2: # apple has gone off screen OR collided
            if collision or collision2:
                if apple["img"] == apple_img_1:
                    if mult_active:
                        score += 1 * multiplier
                    else:
                        score += 1
                elif apple["img"] == apple_img_5:
                    if mult_active:
                        score += 5 * multiplier
                    else:
                        score += 5
                elif apple["img"] == can_img:
                    score -= random.randint(1, 50)
                elif apple["img"] == mult_img:
                    player_speed = 10
                    if mult_active:
                        mult_end_time = time.time() + 5
                        multiplier *= 2
                    else:
                        mult_active = True
                        mult_end_time = time.time() + 5
            apple["y"] = random.randint(-600, -50)
            apple["x"] = random.randint(0, 750)
            # bias towards lower scoring apples
            apple["img"] = random.choice(choices_list)
            
        screen.blit(apple["img"], (apple["x"], apple["y"]))

    screen.blit(player_img, (player_x, player_y))
    if multiplayer:
        screen.blit(player2_img, (player2_x, player2_y))

    score_text = font.render(f"Score: {score}", True, (0, 0, 0))
    screen.blit(score_text, (10, 10))

    timer_text = font.render(f"Score: {time_remaining}s", True, (0, 0, 0))
    screen.blit(timer_text, (10, 50))

    high_score_text = font.render(f"High score: {high_score}", True, (0, 0, 0))
    screen.blit(high_score_text, (10, 90))

    if mult_active:
        mult_text = font.render(f"Multiplier active: x{multiplier}", True, (0, 0, 0))
        screen.blit(mult_text, (10, 130))
    
    pygame.display.flip()

    clock.tick(60)

pygame.quit()
```
