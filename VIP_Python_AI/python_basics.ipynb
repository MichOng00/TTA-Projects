```python
# exercise: simulate rolling a dice until you get 6. Print how many times you had to roll.
# hint: use while loop since we don't know how many times it will take
num_rolls = 0
while True:
    dice = random.randint(1, 6)
    print(dice)
    # num_rolls = num_rolls + 1
    num_rolls += 1 # shortcut for the line above
    if dice == 6:
        print(f"You got 6 in {num_rolls} rolls")
        break
```

```python
# another version with while < condition >
import random
num_rolls = 0
dice = random.randint(1, 6) # must define before using it
while dice != 6:
    dice = random.randint(1, 6)
    print(dice)
    # num_rolls = num_rolls + 1
    num_rolls += 1 # shortcut for the line above
    
print(f"You got 6 in {num_rolls} rolls")
```

```python
# number guessing with multiple guesses
# exercise: print out how many attempts you took in the end
import random
random_number = random.randint(1, 20)
times = 0
max_attempts = 5
while True:
    times += 1
    user_guess = int(input("Guess a number from 1 to 20"))
    print(f"Your guess: {user_guess}")
    if random_number == user_guess:
        print("Correct")
        print(f"You guessed it in {times} attempts")
        break
    elif user_guess > random_number:
        print("Too high")
    else: # don't need user_guess < random_number because that is the only other possibility
        print("Too low")

    if times > max_attempts:
        print("Sorry, you are out of attempts.")
        break
```

```python
# HW version
import random
max_num = int(input("Enter the maximum number"))
random_number = random.randint(1, max_num)
times = 0
max_attempts = 5
while True:
    times += 1
    user_guess = int(input(f"Guess a number from 1 to {max_num}"))
    print(f"Your guess: {user_guess}")
    if random_number == user_guess:
        print("Correct")
        if times == 1:
            print("You guessed it in 1 attempt!!!!")
        else:
            print(f"You guessed it in {times} attempts")
        break
    elif user_guess > random_number:
        print("Too high")
    else:
        print("Too low")

    if times > max_attempts:
        print("Sorry, you are out of attempts.")
        break
```

```python
# example: to-do list
todo = []

while True:
    command = input("Enter a command (add / show / done / quit):")
    if command == "add":
        task = input("Enter the new task:")
        todo.append({"task": task, "done": False})
    elif command == "quit":
        print("Goodbye!")
        break
    elif command == "show":
        if len(todo) == 0:
            print("Nothing to do!")
        else:
            for i, task in enumerate(todo):
                mark = "x" if task["done"] else " "
                print(f"{i+1}. [{mark}] {task["task"]}")
    elif command == "done":
        completed = int(input("Which number task is done?"))
        todo[completed - 1]["done"] = True
    else:
        print("Unknown command.")
```
