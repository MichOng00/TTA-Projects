```python
from ugot import ugot
got = ugot.UGOT()
got.initialize("192.168.1.206")
```

```python
got.mecanum_move_speed_times(0, 20, 50, 1) # forward 50 cm
got.mecanum_move_speed_times(1, 20, 50, 1) # backward 50 cm
```

```python
got.mecanum_turn_speed_times(2, 45, 90, 2) # turn left 90 degrees
got.mecanum_turn_speed_times(3, 45, 90, 2) # turn right 90 degrees
```

```python
got.screen_display_background(2)
```

```python
import time

got.mecanum_move_xyz(20, 0, 0)
time.sleep(1)
got.mecanum_stop()
```

```python
while True:
    dist = got.read_distance_data(51)
    print(dist, end="           \r")
```
