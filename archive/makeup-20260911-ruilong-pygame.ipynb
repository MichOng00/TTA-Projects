```python
import pygame
import random
pygame.init()
screen = pygame.display.set_mode((800, 600))
pygame.display.set_caption("hi")

player_img = pygame.image.load("firefox.jpg")
player_img = pygame.transform.scale(player_img, (100, 100))
enemy_img = pygame.image.load("chrome.jpg")
enemy_img = pygame.transform.scale(enemy_img, (100, 100))
bg = pygame.image.load("google_background.png")
bg = pygame.transform.scale(bg, (800, 600))

x, y = 400, 300
enemy_x, enemy_y = random.randint(0, 700), random.randint(0, 500)
enemy_direction = [random.choice([-1, 1]), random.choice([-1, 1])]
enemy_speed = 0.3
obstacle_x, obstacle_y = random.randint(0, 700), random.randint(0, 500)

hits = 0
font = pygame.font.Font(None, 30)

start_time = pygame.time.get_ticks()

running = True
game_over = False
while running:
    while game_over:
        # screen.fill((255, 255, 255))
        screen.blit(bg, (0,0))
        result_text = font.render("Game over! Press 'R' to play again.", True, (255, 0, 0))
        screen.blit(result_text, (screen.get_width()//2 - result_text.get_width()//2, 100))
        pygame.display.flip()
        for event in pygame.event.get():
            if event.type == pygame.QUIT:
                running = False
                game_over = False
            if event.type == pygame.KEYDOWN:
                if event.key == pygame.K_r:
                    game_over = False
                    x, y = 400, 300
                    enemy_x, enemy_y = random.randint(0, 700), random.randint(0, 500)
                    enemy_direction = [random.choice([-1, 1]), random.choice([-1, 1])]
                    hits = 0
                    start_time = pygame.time.get_ticks()

    elapsed_time = (pygame.time.get_ticks() - start_time) // 1000
    

    if hits >= 3:
        game_over = True
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    keys = pygame.key.get_pressed()
    # print(keys)
    if keys[pygame.K_a]:
        x -= 1
    if keys[pygame.K_d]:
        x += 1
    if keys[pygame.K_w]:
        y -= 1
    if keys[pygame.K_s]:
        y += 1

    if x > 700:
        x = 700
    if x < 0:
        x = 0
    if y > 500:
        y = 500
    if y < 0:
        y = 0

    enemy_x += enemy_direction[0] * enemy_speed
    enemy_y += enemy_direction[1] * enemy_speed

    if enemy_x <= 0 or enemy_x >= 700:
        enemy_direction[0] *= -1
    if enemy_y <= 0 or enemy_y >= 500:
        enemy_direction[1] *= -1

    screen.fill((255, 255, 255))

    screen.blit(player_img, (x, y))
    screen.blit(enemy_img, (enemy_x, enemy_y))

    player_rect = pygame.Rect(x, y, 100, 100)
    pygame.draw.rect(screen, (0, 100, 100), player_rect, 2) # (x, y, width, height)
    enemy_rect = pygame.Rect(enemy_x, enemy_y, 100, 100)
    collision = player_rect.colliderect(enemy_rect)
    if collision:
        hits += 1
        pygame.draw.rect(screen, (255, 100, 100), enemy_rect, 2) # (x, y, width, height)
        pygame.display.flip()
        pygame.time.wait(1000)
        while collision:
            x, y = random.randint(0, 700), random.randint(0, 500)
            player_rect = pygame.Rect(x, y, 100, 100)
            collision = player_rect.colliderect(enemy_rect)
    else:
        pygame.draw.rect(screen, (0, 100, 100), enemy_rect, 2) # (x, y, width, height)

    obstacle_rect = pygame.Rect(obstacle_x, obstacle_y, 100, 100)
    pygame.draw.rect(screen, (0, 100, 100), obstacle_rect, 2)
    obs_collision = obstacle_rect.colliderect(enemy_rect)
    if obs_collision:
        delta_x = obstacle_rect.centerx - enemy_rect.centerx
        delta_y = obstacle_rect.centery - enemy_rect.centery
        if abs(delta_x) > abs(delta_y):
            enemy_direction[0] *= -1
        else:
            enemy_direction[1] *= -1
        obstacle_x, obstacle_y = random.randint(0, 700), random.randint(0, 500)

    hits_text = font.render(f"Hits: {hits}", True, (0,0,0))
    screen.blit(hits_text, (10, 10))
    time_text = font.render(f"Time: {elapsed_time}s", True, (0,0,0))
    screen.blit(time_text, (10, 50))
    pygame.display.flip()


pygame.quit() 
```
