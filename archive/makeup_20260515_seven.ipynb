```python
# comparison operators
print(8 > 7) # more than
print(8 < 7) # less than
print(8 >= 8) # more than OR equal
print(8 <= 8) # less than OR equal
print("apple" == "Apple") # equals (case-sensitive)
print("apple" != "Apple") # not equals
```

```python
# logical operators (combine two Booleans)
# and (returns True only if both sides are true)
print(True and True)
print(True and False)
print(False and False)

# example: watch a movie
has_ticket = True
is_adult = False # for seven
if has_ticket and is_adult:
    print("you can watch demon slayer")
else:
    print("get a ticket / grow older")
```

```python
# or (returns True only if at least one side is True)
print(True or True)
print(True or False)
print(False or False)

# example: buying clothes
my_money = False
FTrue = True
parent_money = FTrue
if my_money or parent_money:
    print("yay new clothes!")
else:
    print("nooooooooo no money to buy clothes")
```

```python
# not (opposite)
print(not True)
print(not False)
print(not not not not not not not not False)

# example: rain
raining = False
if not raining:
    print("you can go out to the suntec city")
else:
    print("stay home! do your homework!")
```

```python
# if-elif-else
a = 5
b = 67
if a > b:
    print("a is bigger than b")
else:
    if a < b:
        print("a is less than b")
    else:
        print("a is equal to b")

# same as above
if a > b:
    print("a is bigger than b")
elif a < b:
    print("a is less than b")
else:
    print("a is equal to b")
```
