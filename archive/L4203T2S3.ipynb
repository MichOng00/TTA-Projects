3. Motion detection robot

```python
# Motion detection
import time
import numpy as np
import cv2
from IPython.display import clear_output,display,Image

got.open_camera()
print("The camera is on.")

while True:
    Laser1 = got.read_distance_data(21)
    picture  = got.read_camera_data()

    if Laser1>30:
        got.show_light_rgb_effect(0,255,0,3)
        time.sleep(0.1)

    else:
        if picture is None:
            print("No data founded.")
            continue
        picture = np.frombuffer(picture, np.uint8)
        picture = cv2.imdecode(picture,cv2.IMREAD_COLOR)
        ok,picture = cv2.imencode(".jpg",picture)
        display(Image(picture.tobytes()))
        got.show_light_rgb_effect(255,0,0,3)
        time.sleep(2)
        got.turn_off_lights()
        break

got.play_audio_tts("Gotcha,ha ha ha ha.", 0, False)
```

4. Spin robotic guard

```python
# Spin Robo Patrol
import time
import numpy as np
import cv2
from IPython.display import,display,Image

got.open_camera()
print("The camera is on.")

while True:
    Laser1 = got.read_distance_data(21)
    picture  = got.read_camera_data()

    if Laser1>50:
        got.mecanum_turn_speed(2,10)
        time.sleep(0.1)

    else:
        if picture is None:
            print("No data founded.")
            continue
        got.mecanum_stop()
        picture = np.frombuffer(picture, np.uint8)
        picture = cv2.imdecode(picture,cv2.IMREAD_COLOR)
        ok,picture = cv2.imencode(".jpg",picture)
        display(Image(picture.tobytes()))
        got.show_light_rgb_effect(255,0,0,3)
        time.sleep(2)
        got.turn_off_lights()
        break

got.play_audio_tts("Gotcha,ha ha ha ha.", 0, False)
```

5. Area sweeping detection

```python
import time
import numpy as np
import cv2
from IPython.display import display, Image

#step 1 define a function
def take_photo():
    picture = got.read_camera_data()
    
    if picture is None:
        print("No camera data found.")
        return
    
    picture = np.frombuffer(picture, np.uint8)
    picture = cv2.imdecode(picture, cv2.IMREAD_COLOR)
    ok, picture = cv2.imencode(".jpg", picture)
    display(Image(picture.tobytes()))
    got.show_light_rgb_effect(255, 0, 0, 3)
    time.sleep(2)
    got.turn_off_lights()
    got.play_audio_tts("Gotcha, ha ha ha ha.", 0, False)

# Step 2: start sweeping
got.open_camera()
print("The camera is on.")

while True:
    for i in range(16):
        Laser1 = got.read_distance_data(21)
        
        if Laser1 > 50:  # No obstacle - keep searching
            got.mecanum_turn_speed(2, 20)  # Counter-clockwise, slow
            got.show_light_rgb_effect(0, 255, 0, 3)
            time.sleep(0.5)
        
        else:  # Obstacle detected!
            got.mecanum_stop()
            take_photo()
            break
    
    if Laser1 <50:
        break
    
    # Sweep RIGHT (clockwise)
    for i in range(16):
        Laser1 = got.read_distance_data(21)
        
        if Laser1 > 50:  # No obstacle - keep searching
            got.mecanum_turn_speed(3, 20)  # Clockwise, slow
            got.show_light_rgb_effect(0, 255, 0, 3)
            time.sleep(0.5)
        
        else:  # Obstacle detected!
            got.mecanum_stop()
            take_photo()
            break
    
    if Laser1 < 50:
        break

got.mecanum_stop()
print("Mission completed.")
```
