```python
from ugot import ugot
got = ugot.UGOT()
got.initialize("192.168.1.xxx")
import time

colors = {"red" : (255, 0, 0),
          "yellow" : (255, 255, 0),
          "blue" : (0, 0, 255), 
          "white" : (255, 255, 255)}

print("what's your favorite color?")

while True:
    question = got.start_audio_asr()
    if not question:
        continue # ignore the rest of the loop, go back to the start

    got.start_audio_nlp(question, False)
    print(question)

    command = question.lower().strip() # make it lowercase and remove spaces
    
    for color, (r, g, b) in colors.items():
        if color in command:
            got.show_light_rgb_effect(r, g, b, 3)
            print(f"I am showing {color} now.")
            time.sleep(3)

    if "stop" in command:
        print("game over")
        got.turn_off_lights()
        break # get out of the loop
```
