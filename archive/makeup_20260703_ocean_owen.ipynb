```python
fruits = ["apple", "banana", "coconut"] # list
print(fruits[0]) # print the first thing in the list
print(fruits[-1]) # print the last thing in the list
fruits.append("dragonfruit") # add something to the end of the list
print(fruits)
fruits.pop() # remove the last thing in the list
print(fruits)
```

```python
# dictionary
# key : value
people = {"ocean" : 10, "owen" : 9, "teacher joshua" : 10}
print(people["owen"]) # use the KEY to get the VALUE
print(people["teacher joshua"]) 

# updating the dictionary
people["teacher joshua"] = 11 # add or change a VALUE
print(people["teacher joshua"])
people.pop("teacher joshua") # remove a KEY and its VALUE from the dictionary
print(people)

# get specific views
print(people.keys()) # get all the names (KEYS)
print(people.values()) # get all the ages (VALUES)
```
