```python
import random
random_number = random.randint(0, 100)

while True:
    guess = int(input("Input a number: "))
    print(guess)
    if guess < random_number:
        print("low")
    elif guess > random_number:
        print("high")
    else:
        print("same")
        break
```

```python
from ugot import ugot
got = ugot.UGOT()
got.initialize("192.168.1.105")
```

```python
got.mecanum_translate_speed_times(90, 30, 50, 1)
```

```python
got.mecanum_turn_speed_times(3, 30, 90, 2)
```

```python
sides = random.randint(3,8)
for i in range(sides):
    got.mecanum_translate_speed_times(0, 30, 50, 1)
    got.mecanum_turn_speed_times(3, 30, int(360/sides), 2)
```

```python
for sides in range(3, 6):
    # sides = j
    for i in range(sides):
        got.mecanum_translate_speed_times(0, 30, 50, 1)
        got.mecanum_turn_speed_times(3, 30, int(360/sides), 2)
```
