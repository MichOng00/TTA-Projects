```python
from ugot import ugot
got = ugot.UGOT()
got.initialize("")

got.load_models(["apriltag_qrcode"])
```

```python
while True:
    try:
        ap_list = got.get_apriltag_total_info()
        print(ap_list)

        if ap_list: # check that the list is not empty
            x = ap_list[0][1]
            distance = ap_list[0][6]

            if x < 280:
                got.mecanum_turn_speed(2, 10)
            elif x > 360:
                got.mecanum_turn_speed(3, 10)
            
            # if the tag is too far, drive forward
                
            # if the tag is too close, drive backward
            
            else:
                got.mecanum_stop()
        else: # if there is no apriltag seen, stop moving
            got.mecanum_stop()
    except KeyboardInterrupt:
        break
```
