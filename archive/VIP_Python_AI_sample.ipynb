### 1.6 Sequential vs nested conditions
When only **one** action should happen, use `elif` (or nesting). Separate `if` statements are all checked, so several can run.

```python
temperature = 35

print("--- Sequential ifs: every true condition runs ---")
if temperature > 30:
    print("It's hot.")
if temperature > 20:
    print("It's warm.")
if temperature > 10:
    print("It's mild.")

print("--- if / elif chain: only one runs ---")
if temperature > 30:
    print("It's hot.")
elif temperature > 20:
    print("It's warm.")
elif temperature > 10:
    print("It's mild.")

print("--- Nested if: a second decision inside the first ---")
raining = True
if temperature > 30:
    if raining:
        print("Hot and rainy: stay in the shade.")
    else:
        print("Hot and dry: bring water.")
else:
    print("Not too hot today.")
```

### 4.6 Mini-project: text-based battle game using functions

```python
import random
import time


def make_character(name, health, attack_power):
    return {"name": name, "health": health, "max_health": health, "attack": attack_power}


def show_status(character):
    print(f"{character['name']}: {character['health']}/{character['max_health']} HP")


def is_alive(character):
    return character["health"] > 0


def attack(attacker, target):
    damage = random.randint(1, attacker["attack"])
    target["health"] = max(0, target["health"] - damage)
    print(f"{attacker['name']} hits {target['name']} for {damage} damage!")


def heal(character):
    amount = random.randint(5, 15)
    character["health"] = min(character["max_health"], character["health"] + amount)
    print(f"{character['name']} heals {amount} HP.")


player = make_character("Hero", 50, 12)
monster = make_character("Goblin", 40, 10)
heals_left = 3

print("A wild Goblin appears!\n")

while is_alive(player) and is_alive(monster):
    show_status(player)
    show_status(monster)
    choice = input("Attack (a) or heal (h)? ").lower()

    if choice == "a":
        attack(player, monster)
    elif choice == "h" and heals_left > 0:
        heal(player)
        heals_left -= 1
        print(f"({heals_left} heals left)")
    elif choice == "h":
        print("No heals left! You lose your turn.")
    else:
        print("Please type a or h.")
        continue

    time.sleep(0.5)
    if is_alive(monster):
        attack(monster, player)
        time.sleep(0.5)
    print()

if is_alive(player):
    print("You defeated the Goblin. You win!")
else:
    print("You were defeated. Game over.")
```

# Part 2: Pygame

### Pygame setup: run this cell first
It imports the modules the games need and defines two helpers:
- `load_image()` loads and scales a picture. If the file is missing, it draws a coloured placeholder so the code still runs.
- `load_sound()` loads a sound file if it exists, otherwise returns `None`.

```python
import os
import random
import time
import pygame


def load_image(filename, size, color=(200, 200, 200), label=""):
    """Load an image and scale it. If the file is missing, draw a coloured placeholder."""
    if os.path.exists(filename):
        img = pygame.image.load(filename)
        return pygame.transform.scale(img, size)
    pygame.font.init()
    surface = pygame.Surface(size, pygame.SRCALPHA)
    pygame.draw.rect(surface, color, (0, 0, size[0], size[1]), border_radius=8)
    if label:
        text = pygame.font.Font(None, 28).render(label, True, (0, 0, 0))
        surface.blit(text, (size[0] // 2 - text.get_width() // 2,
                            size[1] // 2 - text.get_height() // 2))
    return surface


def load_sound(filename):
    """Return a pygame Sound if the file exists and audio works, otherwise None."""
    try:
        if not pygame.mixer.get_init():
            pygame.mixer.init()
        if os.path.exists(filename):
            return pygame.mixer.Sound(filename)
    except pygame.error:
        pass
    return None
```

Remember: the top-left corner of the window is (0, 0). x increases to the right and y increases **downwards**.

### 5.1 Window, game loop and drawing shapes

```python
import pygame

pygame.init()

screen = pygame.display.set_mode((800, 600))
pygame.display.set_caption("hello")

x, y = 100, 100
speed = 1

running = True
while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    screen.fill((255, 255, 255))

    # draw shapes
    # rect (surface, color, position (x, y, width, height))
    pygame.draw.rect(screen, (0, 0, 50), (100, 100, 200, 100))
    # circle (surface, color, centre (x, y), radius)
    pygame.draw.circle(screen, (255, 0, 0), (x, y), 50)
    # polygon (surface, color, points)
    pygame.draw.polygon(screen, (255, 100, 0), [(100, 150), (200, 150), (150, 300)])

    # update display
    pygame.display.flip()

pygame.quit()
```

### 5.2 Mini-project: keyboard control and screen wrapping
The four movement checks are separate `if` statements (not `elif`) so two keys held together give diagonal movement.

```python
import pygame

pygame.init()

screen = pygame.display.set_mode((800, 600))
pygame.display.set_caption("Move and wrap")

x, y = 100, 100
speed = 3

running = True
while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    # held keys: checked every frame
    keys = pygame.key.get_pressed()
    if keys[pygame.K_LEFT]:
        x -= speed
    if keys[pygame.K_RIGHT]:
        x += speed
    if keys[pygame.K_UP]:
        y -= speed
    if keys[pygame.K_DOWN]:
        y += speed

    # wrap around the edges (an object can only leave one side per axis, so elif is fine)
    if x < 0:
        x = 800
    elif x > 800:
        x = 0
    if y < 0:
        y = 600
    elif y > 600:
        y = 0

    screen.fill((255, 255, 255))
    pygame.draw.rect(screen, (0, 0, 50), (100, 100, 200, 100))
    pygame.draw.circle(screen, (255, 0, 0), (x, y), 50)
    pygame.draw.polygon(screen, (255, 100, 0), [(100, 150), (200, 150), (150, 300)])
    pygame.display.flip()

pygame.quit()
```

### 6.1 Sprites, rectangles and collision

```python
pygame.init()

screen = pygame.display.set_mode((800, 600))
pygame.display.set_caption("Sprites and collision")
clock = pygame.time.Clock()

player_img = load_image("zelda.png", (50, 50), (60, 120, 220), "P")
enemy_img = load_image("enemy.png", (50, 50), (220, 60, 60), "E")

player_pos = [100, 100]     # [x, y]
enemy_pos = [300, 300]
speed = 3

running = True
while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    keys = pygame.key.get_pressed()
    if keys[pygame.K_LEFT]:
        player_pos[0] -= speed
    if keys[pygame.K_RIGHT]:
        player_pos[0] += speed
    if keys[pygame.K_UP]:
        player_pos[1] -= speed
    if keys[pygame.K_DOWN]:
        player_pos[1] += speed

    screen.fill((255, 255, 255))
    screen.blit(player_img, player_pos)
    screen.blit(enemy_img, enemy_pos)

    # collision boxes: *player_pos unpacks [x, y] into x, y
    player_rect = pygame.Rect(*player_pos, 50, 50)
    enemy_rect = pygame.Rect(*enemy_pos, 50, 50)

    # visual debugging: width=3 draws only the outline
    pygame.draw.rect(screen, (0, 255, 0), player_rect, width=3)
    pygame.draw.rect(screen, (0, 255, 0), enemy_rect, width=3)

    collision = player_rect.colliderect(enemy_rect)
    if collision:
        pygame.draw.rect(screen, (255, 0, 0), player_rect, width=3)

    pygame.display.flip()
    clock.tick(60)      # limit to 60 frames per second

pygame.quit()
```

### 6.2 Mini-project: player vs stationary enemy with a hit counter
The player is kept on screen (clamping) and each new touch adds one hit. The `was_colliding` flag stops the counter rising every frame while the two boxes overlap.

```python
pygame.init()

screen = pygame.display.set_mode((800, 600))
pygame.display.set_caption("Hit counter")
clock = pygame.time.Clock()
font = pygame.font.SysFont("Arial", 30)

player_img = load_image("zelda.png", (50, 50), (60, 120, 220), "P")
enemy_img = load_image("enemy.png", (50, 50), (220, 60, 60), "E")

player_pos = [100, 100]
enemy_pos = [400, 300]
speed = 4
hits = 0
was_colliding = False

running = True
while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    keys = pygame.key.get_pressed()
    if keys[pygame.K_LEFT]:
        player_pos[0] -= speed
    if keys[pygame.K_RIGHT]:
        player_pos[0] += speed
    if keys[pygame.K_UP]:
        player_pos[1] -= speed
    if keys[pygame.K_DOWN]:
        player_pos[1] += speed

    # clamping: stop the player at the edges instead of wrapping
    if player_pos[0] < 0:
        player_pos[0] = 0
    elif player_pos[0] + 50 > 800:
        player_pos[0] = 800 - 50
    if player_pos[1] < 0:
        player_pos[1] = 0
    elif player_pos[1] + 50 > 600:
        player_pos[1] = 600 - 50

    player_rect = pygame.Rect(*player_pos, 50, 50)
    enemy_rect = pygame.Rect(*enemy_pos, 50, 50)

    collision = player_rect.colliderect(enemy_rect)
    if collision and not was_colliding:     # only count the first frame of a touch
        hits += 1
    was_colliding = collision

    screen.fill((255, 255, 255))
    screen.blit(player_img, player_pos)
    screen.blit(enemy_img, enemy_pos)
    if collision:
        pygame.draw.rect(screen, (255, 0, 0), player_rect, width=3)

    hits_text = font.render(f"Hits: {hits}", True, (0, 0, 0))
    screen.blit(hits_text, (10, 10))

    pygame.display.flip()
    clock.tick(60)

pygame.quit()
```

**Game rules:** survive 30 seconds without being hit 3 times. Press **C** on the start and results screens to play.

Optional sound files: `hit.wav` (plays when hit) and `music.wav` (background music). The game works without them.

```python
pygame.init()

screen = pygame.display.set_mode((800, 600))
pygame.display.set_caption("Dodge the enemy")
clock = pygame.time.Clock()
font = pygame.font.SysFont("Arial", 30)

player_img = load_image("zelda.png", (50, 50), (60, 120, 220), "P")
enemy_img = load_image("enemy.png", (100, 100), (220, 60, 60), "ENEMY")
hit_sound = load_sound("hit.wav")
music = load_sound("music.wav")

player_pos = [100, 100]
enemy_pos = [random.randint(0, 700), random.randint(0, 500)]
# each direction is -1 (left/up) or 1 (right/down)
enemy_direction = [random.choice([-1, 1]), random.choice([-1, 1])]
enemy_speed = 2
speed = 3
hits = 0

game_duration = 30000           # milliseconds
start_time = pygame.time.get_ticks()

running = True
game_over = True                # start on the "Press C" screen
first_round = True

while running:
    # ---------- start / results screen ----------
    while game_over:
        if pygame.mixer.get_init():
            pygame.mixer.stop()
        screen.fill((255, 255, 255))
        if first_round:
            result_text = font.render("Dodge the enemy! Press 'C' to start.", True, (0, 0, 0))
        elif hits < 3:
            result_text = font.render("You win! Press 'C' to play again.", True, (0, 200, 0))
        else:
            result_text = font.render("You lost! Press 'C' to play again.", True, (255, 0, 0))
        # centre the text: screen centre minus half the text width
        screen.blit(result_text, (screen.get_width() // 2 - result_text.get_width() // 2, 200))
        pygame.display.flip()

        for event in pygame.event.get():
            if event.type == pygame.QUIT:
                running = False
                game_over = False
            if event.type == pygame.KEYDOWN:        # one key press, not a held key
                if event.key == pygame.K_c:
                    game_over = False
                    first_round = False
                    # reset the game state
                    player_pos = [100, 100]
                    enemy_pos = [random.randint(0, 700), random.randint(0, 500)]
                    enemy_direction = [random.choice([-1, 1]), random.choice([-1, 1])]
                    hits = 0
                    start_time = pygame.time.get_ticks()
                    if music:
                        music.play(-1)              # -1 means loop forever

    # ---------- one frame of gameplay ----------
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    keys = pygame.key.get_pressed()                 # held keys
    if keys[pygame.K_LEFT]:
        player_pos[0] -= speed
    if keys[pygame.K_RIGHT]:
        player_pos[0] += speed
    if keys[pygame.K_UP]:
        player_pos[1] -= speed
    if keys[pygame.K_DOWN]:
        player_pos[1] += speed

    # enemy moves by direction x speed, then bounces off the edges
    enemy_pos[0] += enemy_direction[0] * enemy_speed
    enemy_pos[1] += enemy_direction[1] * enemy_speed
    if enemy_pos[0] <= 0 or enemy_pos[0] + 100 >= 800:
        enemy_direction[0] *= -1
    if enemy_pos[1] <= 0 or enemy_pos[1] + 100 >= 600:
        enemy_direction[1] *= -1

    # keep the player on screen
    if player_pos[0] < 0:
        player_pos[0] = 0
    elif player_pos[0] + 50 >= 800:
        player_pos[0] = 800 - 50
    if player_pos[1] < 0:
        player_pos[1] = 0
    elif player_pos[1] + 50 >= 600:
        player_pos[1] = 600 - 50

    player_rect = pygame.Rect(*player_pos, 50, 50)
    enemy_rect = pygame.Rect(*enemy_pos, 100, 100)

    screen.fill((255, 255, 255))
    screen.blit(enemy_img, enemy_pos)
    screen.blit(player_img, player_pos)

    if player_rect.colliderect(enemy_rect):
        hits += 1
        if hit_sound:
            hit_sound.play()
        screen.blit(font.render("You got hit!", True, (255, 0, 0)), player_pos)
        pygame.draw.rect(screen, (255, 0, 0), player_rect, width=3)
        pygame.display.flip()
        pygame.time.wait(1000)                      # pause so the player sees the hit
        player_pos = [random.randint(0, 750), random.randint(0, 550)]   # respawn

    # time remaining
    elapsed_time = pygame.time.get_ticks() - start_time
    remaining_time = max(0, (game_duration - elapsed_time) // 1000)

    screen.blit(font.render(f"Hits: {hits}", True, (0, 0, 0)), (10, 10))
    screen.blit(font.render(f"Time: {remaining_time}s", True, (0, 0, 0)), (10, 50))

    if remaining_time == 0 or hits >= 3:
        game_over = True

    pygame.display.flip()
    clock.tick(60)

pygame.quit()
```

**Practice ideas:** make the enemy a little faster each time it bounces (`enemy_speed += 0.1`) · change the number of lives from 3 · add a second enemy · play a different sound for winning and losing.

## Session 8: Catch game

**Topics:** multiple falling objects stored as a list of dictionaries · list comprehension to create objects · per-object collision, respawning and scoring · weighted random selection · timer helper function

**Game rules:** move the basket with the left and right arrow keys. Catch 1-point and 5-point apples for 60 seconds. The 5-point apples are rarer.

```python
pygame.init()

screen = pygame.display.set_mode((800, 600))
pygame.display.set_caption("Catch game")

player_img = load_image("basket.png", (100, 50), (150, 100, 50), "basket")
apple_img_1 = load_image("1apple.png", (50, 50), (220, 40, 40), "1")
apple_img_5 = load_image("5apple.png", (50, 50), (240, 190, 40), "5")

player_x = 350
player_y = 500
score = 0
start_time = time.time()
duration = 60


def get_time_remaining():
    elapsed_time = time.time() - start_time
    return max(0, int(duration - elapsed_time))


clock = pygame.time.Clock()

# a list of dictionaries, built with a list comprehension
apples = [
    {"img": random.choice([apple_img_1, apple_img_5]),
     "x": random.randint(0, 750),
     "y": random.randint(-600, -50),      # start above the screen
     "speed": random.randint(3, 7)}
    for _ in range(5)
]

font = pygame.font.Font(None, 36)

running = True
while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    keys = pygame.key.get_pressed()
    if keys[pygame.K_LEFT] and player_x > 0:
        player_x -= 7
    if keys[pygame.K_RIGHT] and player_x < 800 - 100:
        player_x += 7

    time_remaining = get_time_remaining()
    if time_remaining == 0:
        print(f"Your final score is {score}")
        running = False

    screen.fill((255, 255, 255))
    player_rect = pygame.Rect(player_x, player_y, 100, 50)

    # handle every apple
    for apple in apples:
        apple["y"] += apple["speed"]
        apple_rect = pygame.Rect(apple["x"], apple["y"], 50, 50)
        collision = player_rect.colliderect(apple_rect)

        if apple["y"] > 600 or collision:       # off the bottom OR caught
            if collision:
                if apple["img"] == apple_img_1:
                    score += 1
                elif apple["img"] == apple_img_5:
                    score += 5
            # respawn above the screen with a new image
            apple["y"] = random.randint(-600, -50)
            apple["x"] = random.randint(0, 750)
            # weighted random: 1-point apples appear 3 times in the list, 5-point once
            apple["img"] = random.choice([apple_img_1, apple_img_1, apple_img_1, apple_img_5])

        screen.blit(apple["img"], (apple["x"], apple["y"]))

    screen.blit(player_img, (player_x, player_y))
    score_text = font.render(f"Score: {score}", True, (0, 0, 0))
    screen.blit(score_text, (10, 10))
    timer_text = font.render(f"Time: {time_remaining}s", True, (0, 0, 0))
    screen.blit(timer_text, (10, 50))

    pygame.display.flip()
    clock.tick(60)

pygame.quit()
```

**Practice ideas:** add a 10-point golden apple that appears very rarely · make apples fall faster as the score rises · change the number of apples from 5.

## Session 9: Power-ups and high scores

**Topics:** penalty items and a temporary score multiplier with a timer · building weighted lists with loops · file I/O (`open`, read and write, `with`, `os.path.exists`) · saving and loading a high score

**New items:** an explosion (loses 1–50 points) and a green apple (doubles scores for 5 seconds). The best score is saved in `highscores.txt` next to the notebook.

```python
pygame.init()

screen = pygame.display.set_mode((800, 600))
pygame.display.set_caption("Catch game")

player_img = load_image("basket.png", (100, 50), (150, 100, 50), "basket")
apple_img_1 = load_image("1apple.png", (50, 50), (220, 40, 40), "1")
apple_img_5 = load_image("5apple.png", (50, 50), (240, 190, 40), "5")
enemy_img = load_image("explosion.png", (50, 50), (60, 60, 60), "BOOM")
mult_img = load_image("green_apple.jpg", (50, 50), (60, 200, 80), "x2")

player_x = 350
player_y = 500
score = 0
start_time = time.time()
duration = 60
high_score_file = "highscores.txt"

mult_active = False
mult_end_time = 0
multiplier = 2


def get_time_remaining():
    elapsed_time = time.time() - start_time
    return max(0, int(duration - elapsed_time))


clock = pygame.time.Clock()

# build the weighted list with a loop: 1 enemy, 1 multiplier, 3 of each apple
choices_list = []
choices_list.append(enemy_img)
choices_list.append(mult_img)
for _ in range(3):
    choices_list.append(apple_img_1)
    choices_list.append(apple_img_5)

apples = [
    {"img": random.choice(choices_list),
     "x": random.randint(0, 750),
     "y": random.randint(-600, -50),
     "speed": random.randint(3, 7)}
    for _ in range(5)
]


# high score helpers
def load_high_score():
    if os.path.exists(high_score_file):
        with open(high_score_file, "r") as file:        # r for read
            return int(file.read().strip() or 0)
    return 0


def save_high_score(score):
    with open(high_score_file, "w") as file:            # w for write
        file.write(str(score))


high_score = load_high_score()
font = pygame.font.Font(None, 36)

running = True
while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    keys = pygame.key.get_pressed()
    if keys[pygame.K_LEFT] and player_x > 0:
        player_x -= 7
    if keys[pygame.K_RIGHT] and player_x < 800 - 100:
        player_x += 7

    time_remaining = get_time_remaining()

    # the multiplier switches itself off when its time is up
    if mult_active and time.time() >= mult_end_time:
        mult_active = False

    if time_remaining == 0:
        print(f"Your final score is {score}")
        if score > high_score:
            high_score = score
            save_high_score(high_score)
            print("New high score!")
        running = False

    screen.fill((255, 255, 255))
    player_rect = pygame.Rect(player_x, player_y, 100, 50)

    for apple in apples:
        apple["y"] += apple["speed"]
        apple_rect = pygame.Rect(apple["x"], apple["y"], 50, 50)
        collision = player_rect.colliderect(apple_rect)

        if apple["y"] > 600 or collision:
            if collision:
                if apple["img"] == apple_img_1:
                    if mult_active:
                        score += 1 * multiplier
                    else:
                        score += 1
                elif apple["img"] == apple_img_5:
                    if mult_active:
                        score += 5 * multiplier
                    else:
                        score += 5
                elif apple["img"] == enemy_img:
                    score -= random.randint(1, 50)          # penalty
                elif apple["img"] == mult_img:
                    mult_active = True
                    mult_end_time = time.time() + 5         # lasts 5 seconds
            apple["y"] = random.randint(-600, -50)
            apple["x"] = random.randint(0, 750)
            apple["img"] = random.choice(choices_list)

        screen.blit(apple["img"], (apple["x"], apple["y"]))

    screen.blit(player_img, (player_x, player_y))
    screen.blit(font.render(f"Score: {score}", True, (0, 0, 0)), (10, 10))
    screen.blit(font.render(f"Time: {time_remaining}s", True, (0, 0, 0)), (10, 50))
    screen.blit(font.render(f"High score: {high_score}", True, (0, 0, 0)), (10, 90))
    if mult_active:
        screen.blit(font.render(f"Multiplier active: x{multiplier}", True, (0, 0, 0)), (10, 130))

    pygame.display.flip()
    clock.tick(60)

pygame.quit()
```

**Practice ideas:** keep a list of the best 3 scores instead of one · make the multiplier last 8 seconds · add a "freeze" item that stops the apples for 3 seconds.

## Session 10: Menu, multiplayer and final showcase

**Topics:** menu screen and game states · `.copy()` for resetting game state · two-player controls (arrows and WASD) · stacking power-ups · personalising the final game, demo and peer feedback

**Controls:** player 1 uses the **left/right arrows**, player 2 uses **A/D**. Press **N** for single player or **M** for multiplayer on the menu. Catching green apples gives a speed boost and the multiplier doubles each time you catch another while it is active (x2, x4, x8…).

**Teacher note on `.copy()`:** `apples.copy()` copies only the list. The dictionaries inside are still shared, so changing an apple would change the "saved" copy too. To save a real snapshot, copy each dictionary: `[apple.copy() for apple in apples]`, as the code below does.

```python
pygame.init()

screen = pygame.display.set_mode((800, 600))
pygame.display.set_caption("Catch game")

player_img = load_image("basket.png", (100, 50), (150, 100, 50), "P1")
player2_img = load_image("basket.png", (100, 50), (50, 100, 180), "P2")
apple_img_1 = load_image("1apple.png", (50, 50), (220, 40, 40), "1")
apple_img_5 = load_image("5apple.png", (50, 50), (240, 190, 40), "5")
enemy_img = load_image("explosion.png", (50, 50), (60, 60, 60), "BOOM")
mult_img = load_image("green_apple.jpg", (50, 50), (60, 200, 80), "x2")

player_speed = 7
player_x = 350
player_y = 500
player2_x = 400
player2_y = 500
score = 0
last_score = None
start_time = time.time()
duration = 60
high_score_file = "highscores.txt"

mult_active = False
mult_end_time = 0
multiplier = 2
multiplayer = False


def get_time_remaining():
    elapsed_time = time.time() - start_time
    return max(0, int(duration - elapsed_time))


clock = pygame.time.Clock()

choices_list = []
choices_list.append(enemy_img)
choices_list.append(mult_img)
for _ in range(3):
    choices_list.append(apple_img_1)
    choices_list.append(apple_img_5)

apples = [
    {"img": random.choice(choices_list),
     "x": random.randint(0, 750),
     "y": random.randint(-600, -50),
     "speed": random.randint(3, 7)}
    for _ in range(5)
]
# snapshot of the starting state: copy each dictionary so the snapshot stays unchanged
apples_copy = [apple.copy() for apple in apples]


def load_high_score():
    if os.path.exists(high_score_file):
        with open(high_score_file, "r") as file:
            return int(file.read().strip() or 0)
    return 0


def save_high_score(score):
    with open(high_score_file, "w") as file:
        file.write(str(score))


high_score = load_high_score()
font = pygame.font.Font(None, 36)

game_over = True        # True means "show the menu"
running = True
while running:
    # ---------- menu ----------
    while game_over:
        screen.fill((255, 255, 255))
        menu_text = font.render("Press N for single player, M for multiplayer", True, (0, 0, 0))
        screen.blit(menu_text, (screen.get_width() // 2 - menu_text.get_width() // 2, 200))
        if last_score is not None:
            last_text = font.render(f"Last score: {last_score}", True, (0, 0, 0))
            screen.blit(last_text, (screen.get_width() // 2 - last_text.get_width() // 2, 250))
        pygame.display.flip()

        for event in pygame.event.get():
            if event.type == pygame.QUIT:
                running = False
                game_over = False
            if event.type == pygame.KEYDOWN:
                if event.key == pygame.K_n or event.key == pygame.K_m:
                    multiplayer = (event.key == pygame.K_m)
                    # reset everything for a fresh round
                    score = 0
                    player_x = 350
                    player2_x = 400
                    player_speed = 7
                    multiplier = 2
                    mult_active = False
                    apples = [apple.copy() for apple in apples_copy]
                    start_time = time.time()
                    game_over = False

    # ---------- one frame of gameplay ----------
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    keys = pygame.key.get_pressed()
    if keys[pygame.K_LEFT] and player_x > 0:
        player_x -= player_speed
    if keys[pygame.K_RIGHT] and player_x < 800 - 100:
        player_x += player_speed
    if multiplayer:
        if keys[pygame.K_a] and player2_x > 0:
            player2_x -= player_speed
        if keys[pygame.K_d] and player2_x < 800 - 100:
            player2_x += player_speed

    time_remaining = get_time_remaining()

    # all power-up effects end together
    if mult_active and time.time() >= mult_end_time:
        mult_active = False
        multiplier = 2
        player_speed = 7

    if time_remaining == 0:
        print(f"Your final score is {score}")
        last_score = score
        if score > high_score:
            high_score = score
            save_high_score(high_score)
            print("New high score!")
        game_over = True        # back to the menu instead of quitting

    screen.fill((255, 255, 255))
    player_rect = pygame.Rect(player_x, player_y, 100, 50)
    if multiplayer:
        player2_rect = pygame.Rect(player2_x, player2_y, 100, 50)
    else:
        # dummy rectangle on top of player 1, so it never adds a second collision
        player2_rect = pygame.Rect(player_x, player_y, 100, 50)

    for apple in apples:
        apple["y"] += apple["speed"]
        apple_rect = pygame.Rect(apple["x"], apple["y"], 50, 50)
        collision = player_rect.colliderect(apple_rect)
        collision2 = player2_rect.colliderect(apple_rect)

        if apple["y"] > 600 or collision or collision2:
            if collision or collision2:
                if apple["img"] == apple_img_1:
                    if mult_active:
                        score += 1 * multiplier
                    else:
                        score += 1
                elif apple["img"] == apple_img_5:
                    if mult_active:
                        score += 5 * multiplier
                    else:
                        score += 5
                elif apple["img"] == enemy_img:
                    score -= random.randint(1, 50)
                elif apple["img"] == mult_img:
                    player_speed = 10                       # speed boost
                    if mult_active:
                        mult_end_time = time.time() + 5     # extend the time
                        multiplier *= 2                     # stack: x2, x4, x8...
                    else:
                        mult_active = True
                        mult_end_time = time.time() + 5
            apple["y"] = random.randint(-600, -50)
            apple["x"] = random.randint(0, 750)
            apple["img"] = random.choice(choices_list)

        screen.blit(apple["img"], (apple["x"], apple["y"]))

    screen.blit(player_img, (player_x, player_y))
    if multiplayer:
        screen.blit(player2_img, (player2_x, player2_y))

    screen.blit(font.render(f"Score: {score}", True, (0, 0, 0)), (10, 10))
    screen.blit(font.render(f"Time: {time_remaining}s", True, (0, 0, 0)), (10, 50))
    screen.blit(font.render(f"High score: {high_score}", True, (0, 0, 0)), (10, 90))
    if mult_active:
        screen.blit(font.render(f"Multiplier active: x{multiplier}", True, (0, 0, 0)), (10, 130))

    pygame.display.flip()
    clock.tick(60)

pygame.quit()
```
