```python
import pygame

pygame.init()
screen = pygame.display.set_mode((800, 600))

x, y = 400, 300
x2, y2 = 400, 300

running = True
while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    keys = pygame.key.get_pressed()
    if keys[pygame.K_a]:
        x -= 1
    if keys[pygame.K_d]:
        x += 1
    if keys[pygame.K_w]:
        y -= 1
    if keys[pygame.K_s]:
        y += 1

    if keys[pygame.K_LEFT]:
        x2 -= 1
    if keys[pygame.K_RIGHT]:
        x2 += 1
    if keys[pygame.K_UP]:
        y2 -= 1
    if keys[pygame.K_DOWN]:
        y2 += 1

    screen.fill((255, 255, 255))
    pygame.draw.circle(screen, (0, 100, 100), (x, y), 50)
    pygame.draw.circle(screen, (200, 0, 100), (x2, y2), 50)
    pygame.display.flip()

pygame.quit()
```
