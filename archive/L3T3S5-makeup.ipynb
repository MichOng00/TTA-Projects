```python
from ugot import ugot
got = ugot.UGOT()
got.initialize("192.168.1.200")
```

```python
got.mechanical_joint_control(0, 0, 0, 500)
```

```python
import time

got.mechanical_clamp_release() # open gripper
time.sleep(1)
got.mechanical_clamp_close() # close gripper
```

```python
# defining functions
def add(num1, num2): # two parameters
    print(num1 + num2)

def say_hello(name): # one parameter
    print(f"hello, {name}")

your_name = input("What is your name? ")
say_hello(your_name)
add(2, 3)
```

```python
# function to pick up
def pickup():
    got.mechanical_clamp_release()
    time.sleep(1)
    got.mechanical_joint_control(0, -35, -10, 500)
    time.sleep(1)
    got.mechanical_clamp_close()
    time.sleep(1)
    got.mechanical_joint_control(0, 0, 0, 500)

def put():
    got.mechanical_joint_control(0, -35, -10, 500)
    time.sleep(1)
    got.mechanical_clamp_release()
```

```python
pickup()
got.mecanum_translate_speed_times(0, 30, 50, 1)
put()
got.mecanum_turn_speed_times(2, 30, 180, 2)
got.mecanum_translate_speed_times(0, 30, 50, 1)
pickup()
```

```python
# write a program to only pick up when it recognises the correct apriltag id
got.mechanical_joint_control(0, 0, 0, 500)
while True:
    info = got.get_apriltag_total_info()
    ap_id = -1
    if info != []: # check that info is NOT equal to the empty list
        ap_id = info[0][0]
    if ap_id == 6:
        pickup()
        break # get out of the loop

print("picked up")
```
