```python
from ugot import ugot
got = ugot.UGOT()
got.initialize("192.168.1.191")

import time
```

```python
got.mechanical_arms_restory() # reset arm
got.mechanical_clamp_release() # open gripper
time.sleep(1)
got.mechanical_clamp_close() # close gripper
```

```python
got.mechanical_joint_control(angle1=0, angle2=0, angle3=0, duration=500)
```

```python
def pickup(y):
    got.mechanical_clamp_release()
    got.mechanical_joint_control(angle1=0, angle2=-45, angle3=-45, duration=y)
    time.sleep(y / 1000)  # convert y to seconds from milliseconds
    got.mechanical_clamp_close()
    time.sleep(1)
    got.mechanical_joint_control(angle1=0, angle2=0, angle3=0, duration=y)

pickup(500)

def throw():
    got.mechanical_joint_control(angle1=0, angle2=45, angle3=45, duration=500)
    time.sleep(1)
    got.mechanical_joint_control(angle1=0, angle2=0, angle3=0, duration=50)
    got.mechanical_clamp_release()

throw()
```

```python
got.read_distance_data(21) 
# 2: port that the wire is connected to
# 1: first thing connected to that port
```

```python
# function which takes one parameter (threshold) and 
# returns True if there is an obstacle closer than the threshold
def check_distance(threshold):
    if threshold > got.read_distance_data(21):
        return True
    else:
        return False

check_distance(threshold=15)
```

```python
while not check_distance(15):
    got.mecanum_move_speed(0, 10)
got.mecanum_stop()
pickup(500)
throw()
```

```python
# Wordle
secret_word = "robot" 
guessed = False # flag to say whether or not you have guessed the word

while not guessed:
    result = "" # store and print out wordle color boxes
    player_guess = got.get_words_result().lower().strip()
    if len(secret_word) == len(player_guess):
        print(player_guess)
        for i in range(len(secret_word)):
            # correct letter in correct position
            if secret_word[i] == player_guess[i]:
                got.show_light_rgb_effect(0, 255, 0, 0) # green
                result += "🟩"
            # correct letter in wrong position
            elif player_guess[i] in secret_word:
                got.show_light_rgb_effect(255, 255, 0, 0) # yellow
                result += "🟨"
            else:
                got.show_light_rgb_effect(255, 0, 0, 0) # red
                result += "⬛"
            time.sleep(1)
            got.turn_off_lights()
            time.sleep(1)
        print(result)
        if player_guess == secret_word:
            guessed = True
```

```python
import random

words = ["python", "bottle", "laptop", "crane", "avocado"]
secret_word = random.choice(words) # randomly pick an item from a list
display_word = ["_"] * len(secret_word)
# print(display_word)
guessed_letters = []

max_attempts = 6
wrong_attempts = 0

print("Welcome to Python hangman!")
while "_" in display_word and wrong_attempts < max_attempts:
    print(f"\nWord to guess: {" ".join(display_word)}")
    print(f"Attempts remaining: {max_attempts - wrong_attempts}")

    guess = input("Guess a letter").lower()

    if not guess.isalpha() or len(guess) != 1:
        print("Please enter a single letter guess")
        continue  # skip the rest of the loop

    # make sure it is a new guess
    if guess in guessed_letters:
        print(f"You already guessed {guess}!")
        continue

    guessed_letters.append(guess)

    # check if letter is in word
    if guess in secret_word:
        print("correct")
        # update display word
        for i in range(len(secret_word)):
            if guess == secret_word[i]:
                display_word[i] = guess
    else:
        print(f"sorry, '{guess}' is not in the word")
        wrong_attempts += 1

if "_" not in display_word:
    print(f"Congratulations! You guessed the word {secret_word}")
else:
    print(f"You lose! The word was {secret_word}")
```

```python
import numpy as np # numerical python
import cv2         # computer vision

from ugot import ugot
got = ugot.UGOT()
got.initialize("192.168.1.171")
got.load_models(["color_recognition"])
got.open_camera()

frame = got.read_camera_data()
print(frame)
nparr = np.frombuffer(frame, np.uint8) # unsigned integer 8 bit
print(nparr)
data = cv2.imdecode(nparr, cv2.IMREAD_COLOR)
print(data)

from IPython.display import Image, display
display(Image(frame))
```

```python
while True:
    frame = got.read_camera_data()
    nparr = np.frombuffer(frame, np.uint8)
    data = cv2.imdecode(nparr, cv2.IMREAD_COLOR)

    color_info = got.get_color_total_info()
    color = color_info[0]
    shape = color_info[1]

    cv2.putText(data, f"{color} {shape}", (20, 20), 1, 4, (0, 255, 255), 3, cv2.LINE_AA)
    # exercise: move the text to the bottom of the window
    # change the color of the text
    # change the size of the text
    cv2.circle(data, (600, 400), 5, (0, 0, 255), -1, cv2.LINE_AA)

    cv2.imshow("UGOT camera", data) # show in a separate window

    # exit loop when Q is pressed
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cv2.destroyAllWindows()
```
