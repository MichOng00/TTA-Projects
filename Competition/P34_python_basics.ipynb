```python
# number guessing game with while loop
import random
secret_number = random.randint(1, 100)

chances = 8
while chances > 0: # keep guessing until you are out of chances
    chances -= 1 # minus one from the current value of chances
    your_guess = int(input("guess a number from 1 to 5"))
    if your_guess == secret_number:
        print("yay")
        break # exit the loop
    elif your_guess > secret_number:
        print(f"{your_guess} is too high")
    else:
        print(f"{your_guess} is too low")
```

```python
from ugot import ugot
got = ugot.UGOT()
got.initialize("192.168.1.191")

import time
```

```python
got.mechanical_arms_restory() # reset arm
got.mechanical_clamp_release() # open gripper
time.sleep(1)
got.mechanical_clamp_close() # close gripper
```

```python
got.mechanical_joint_control(angle1=0, angle2=0, angle3=0, duration=500)
```

```python
def pickup(y):
    got.mechanical_clamp_release()
    got.mechanical_joint_control(angle1=0, angle2=-45, angle3=-45, duration=y)
    time.sleep(y / 1000)  # convert y to seconds from milliseconds
    got.mechanical_clamp_close()
    time.sleep(1)
    got.mechanical_joint_control(angle1=0, angle2=0, angle3=0, duration=y)

pickup(500)

def throw():
    got.mechanical_joint_control(angle1=0, angle2=45, angle3=45, duration=500)
    time.sleep(1)
    got.mechanical_joint_control(angle1=0, angle2=0, angle3=0, duration=50)
    got.mechanical_clamp_release()

throw()
```

```python
got.read_distance_data(21) 
# 2: port that the wire is connected to
# 1: first thing connected to that port
```

```python
# function which takes one parameter (threshold) and 
# returns True if there is an obstacle closer than the threshold
def check_distance(threshold):
    if threshold > got.read_distance_data(21):
        return True
    else:
        return False

check_distance(threshold=15)
```

```python
while not check_distance(15):
    got.mecanum_move_speed(0, 10)
got.mecanum_stop()
pickup(500)
throw()
```

```python
# Wordle
secret_word = "robot" 
guessed = False # flag to say whether or not you have guessed the word

while not guessed:
    result = "" # store and print out wordle color boxes
    player_guess = got.get_words_result().lower().strip()
    if len(secret_word) == len(player_guess):
        print(player_guess)
        for i in range(len(secret_word)):
            # correct letter in correct position
            if secret_word[i] == player_guess[i]:
                got.show_light_rgb_effect(0, 255, 0, 0) # green
                result += "🟩"
            # correct letter in wrong position
            elif player_guess[i] in secret_word:
                got.show_light_rgb_effect(255, 255, 0, 0) # yellow
                result += "🟨"
            else:
                got.show_light_rgb_effect(255, 0, 0, 0) # red
                result += "⬛"
            time.sleep(1)
            got.turn_off_lights()
            time.sleep(1)
        print(result)
        if player_guess == secret_word:
            guessed = True
```
