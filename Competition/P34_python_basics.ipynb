```python
# comparison operators - give you a boolean
print(2 < 4) # less than
print(2 > 4) # more than
print(2 == 4) # equal
print(2 != 4) # not equal
print(2 >= 4) # more than OR equal
print(2 <= 4) # less than OR equal

print("Jolie" == "Oliver")
print("Ze Sen" == "Ze sen") # case-sensitive
```
