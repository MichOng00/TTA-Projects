```python
# number guessing game
import random
secret_number = random.randint(1, 5)
your_guess = int(input("guess a number from 1 to 5"))

if your_guess == secret_number:
    print("yay")
elif your_guess > secret_number:
    print("too high")
else:
    print("too low")
```

```python
# while loop example
password = "676767"
user_input = None
while password != user_input:
    user_input = input("what is the password?")
    print(user_input)
print("welcome")
```

```python
# number guessing game with while loop
import random
secret_number = random.randint(1, 100)

chances = 8
while chances > 0: # keep guessing until you are out of chances
    chances -= 1 # minus one from the current value of chances
    your_guess = int(input("guess a number from 1 to 5"))
    if your_guess == secret_number:
        print("yay")
        break # exit the loop
    elif your_guess > secret_number:
        print(f"{your_guess} is too high")
    else:
        print(f"{your_guess} is too low")
```

```python
from ugot import ugot
got = ugot.UGOT()
got.initialize("192.168.1.xxx")
got.show_light_rgb_effect(0, 255, 0, 0)

num_sides = 3
degrees = int(360 / num_sides)

for i in range(num_sides):
    # go forward
    got.mecanum_move_speed_times(direction=0, speed=20, times=30, unit=1)
    # turn
    got.mecanum_turn_speed_times(turn=3, speed=40, times=degrees, unit=2)
```

```python
import time
got.mecanum_move_speed(direction=0, speed=20)
time.sleep(2) # wait for 2 seconds
got.mecanum_stop()
```

```python
dist = got.read_distance_data(21)

# exercise: move forward until it senses an obstacle, then stop
while dist > 20:
    dist = got.read_distance_data(21) # update distance reading
    got.mecanum_move_speed(direction=0, speed=20)
got.mecanum_stop()
```
