```python
from ugot import ugot
got = ugot.UGOT()
got.initialize("192.168.1.235")

# use a for loop to drive in a polygon
num_sides = 3
degree = int(360 / num_sides)
for i in range(num_sides):
    got.mecanum_move_speed_times(0, 20, 30, 1)
    got.mecanum_turn_speed_times(2, 45, degree, 2)
```

```python
dist = got.read_distance_data(21) # 2 - port number, 1 - first thing connected to that port

# move forward until it sees an obstacle, then stop
while dist > 20:
    got.mecanum_move_speed(0, 20)
    dist = got.read_distance_data(21) # update reading
got.mecanum_stop()
```
