```python
# connect to ugot
from ugot import ugot
got = ugot.UGOT()
got.initialize("192.168.1.215")
import time

# show light - please use different colours
got.show_light_rgb_effect(255, 0, 0, 3) # red, blue, green, effect
```

```python
# self-balancing basic movement
got.balance_start_balancing()
# move forward
got.balance_move_speed_times(0, 30, 30, 1)
# move backward
got.balance_move_speed_times(1, 30, 30, 1)
# turn left
got.balance_turn_speed_times(2, 50, 90, 2)
# turn right
got.balance_turn_speed_times(3, 50, 90, 2)

# engineer / mecanum basic movement
got.mecanum_turn_speed_times(2, 30, 90, 2)
got.mecanum_move_speed_times(0, 30, 30, 1)
```

```python
# arm control - engineer only
got.mechanical_arms_restory() # reset
got.mechanical_clamp_release()
time.sleep(1)
got.mechanical_joint_control(0, 0, -70, 800)
time.sleep(1)
got.mechanical_clamp_close()
got.mechanical_joint_control(0, 0, 50, 800)
```

```python
# open camera
import numpy as np
import cv2

while True:
    frame = got.read_camera_data()
    nparr = np.frombuffer(frame, np.uint8)
    data = cv2.imdecode(nparr, cv2.IMREAD_COLOR)

    cv2.imshow("UGOT camera", data)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cv2.destroyAllWindows()
```
