```python
import torch
import torch.nn as nn

x = torch.zeros(1, 1, 6, 6) # all zeros 6x6 array (all-dark square)
x[:, :, 2:4, 2:4] = 1.0     # set the centre to 1 (bright)
print("Input:\n", x[0,0])

edge = torch.tensor([[[[-1, -1, -1],
                       [-1, 8, -1],
                       [-1, -1, -1]]]], dtype=torch.float32)
conv = nn.Conv2d(in_channels=1, out_channels=1, kernel_size=3, padding=1, bias=False)
with torch.no_grad():
    conv.weight.copy_(edge)
y = conv(x)
print("Output:\n", y[0,0])

relu = nn.ReLU()
yr = relu(y)
print("ReLU(y)\n:", yr[0,0])

pool = nn.MaxPool2d(2)
yp = pool(yr)
print("Pooled:\n", yp[0,0])
```
