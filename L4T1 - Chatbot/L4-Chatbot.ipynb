```python
# write a calculator function
# using four separate functions for addition, subtraction, division, multiplication
def add(a, b):
    return a + b
def sub(a, b):
    return a - b
def div(a, b):
    if b != 0:
        return a / b
    else:
        return "Error: division by zero"
def mul(a, b):
    return a * b

def calculator(a, b, operation):
    if operation == "+":
        return add(a, b)
    elif operation == "-":
        return sub(a, b)
    elif operation == "*":
        return mul(a, b)
    elif operation == "/":
        return div(a, b)
    else:
        return "invalid operation"
    
print(calculator(2, 3, "+"))
```

## Session 5 - Textblob

```python
from textblob import TextBlob

def analyse_sentiment(text):
    blob = TextBlob(text) # create a textblob object from a string
    return blob.sentiment.polarity

analyse_sentiment("I'm so happy!!!!") # sentiment 1.0 - very positive
analyse_sentiment("the sky is blue") # sentiment 0.0 - neutral
analyse_sentiment("wish you a very merry christmas!!") # punctuation affects score
analyse_sentiment("I'm sad!!!!") # sentiment -1.0 - very negative
```

```python
from ugot import ugot
got = ugot.UGOT()
got.initialize("192.168.1.29")
import random

def recommendation_chatbot():
    activities = {
        # create as many keys as you have cases below
        "very positive" : ["play game", "read"],
        "positive" : ["sing", "exercise"],
        "neutral" : ["sleep", "listen to music"],
        "sad" : ["pet a dog", "pet a cat"]
    }

    print("How are you feeling today?")
    user_response = input("You: ")

    sentiment = analyse_sentiment(user_response)
    if sentiment > 0.5:
        print("Chatbot: you seem really happy!")
        got.screen_display_emotion("Smile")
        # activity = activities["very positive"][random.randint(0, len(activities)-1)]
        activity = random.choice(activities["very positive"])
    elif sentiment > 0:
        print("Chatbot: you seem to be doing well")
        got.screen_display_emotion("WakeUp")
        activity = random.choice(activities["positive"])
    elif sentiment > -0.05:
        print("Chatbot: neutral mood")
        got.screen_display_emotion("Doubt")
        activity = random.choice(activities["neutral"])
    else:
        print("Chatbot: Sounds like you're a little sad.")
        got.screen_display_emotion("Sleep")
        activity = random.choice(activities["sad"])
    print(f"You could do {activity}")

recommendation_chatbot()
```

```python
def error_checking_chatbot():
    """Take user input. Ask if corrected input is what you meant.
        If yes, return it. Otherwise, ask for input again."""
    while True:
        user_input = input()
        blob = TextBlob(user_input)
        corrected = str(blob.correct())
        confirmation = input(f"Did you mean: {corrected}?")
        if confirmation == "yes":
            return corrected
    
error_checking_chatbot()
```

```python
# functions for drawing with turtle and robot
def draw_with_turtle(commands):
    make_turtle(width=800, height=800)
    for cmd, value in commands:
        if cmd == "forward":
            forward(value)
        elif cmd == "back":
            back(value)
        elif cmd == "left":
            left(value)
        elif cmd == "right":
            right(value)
        else:
            print(f"unknown command: {cmd}")

def draw_with_robot(commands, got=got):
    got.balance_start_balancing()
    for cmd, value in commands:
        if cmd == "forward":
            got.balance_move_speed_times(0, 30, value, 1)
        elif cmd == "back":
            got.balance_move_speed_times(1, 30, value, 1)
        elif cmd == "left":
            got.balance_turn_speed_times(3, 30, value, 2)
        elif cmd == "right":
            got.balance_turn_speed_times(2, 30, value, 2)
        else:
            print(f"unknown command: {cmd}")

commands = [["forward", 10], ["right", 50]]
draw_with_turtle(commands)
draw_with_robot(commands)
```

```python
# draw the same shape with both turtle and UGOT
def ugot_follow_turtle():
    while True:
        user_input = input("Enter commands (e.g. forward 50, right 90, exit): ")
        raw_commands = user_input.split(",") # ["forward 50", " right 90"]
        commands = []
        for item in raw_commands:
            parts = item.strip().split(" ")
            if len(parts) == 2:
                cmd = parts[0]
                val = int(parts[1])
                commands.append((cmd, val))
            else:
                if parts[0] == "exit":
                    break
        draw_with_turtle(commands)
        draw_with_robot(commands)

ugot_follow_turtle()
```

```python
def random_painting(steps, length, shape_sides, width, height):
    make_turtle(delay=0.02, width=width, height=height)
    angles = range(0, 361, int(360/shape_sides))
    for i in range(steps):
        set_color(random.choice(colors))
        forward(length)
        # limit turtle to window
        x, y = get_turtle().position
        if x < 0:
            x = 0
        elif x > width:
            x = width
        if y < 0:
            y = 0
        elif y > height:
            y = height
        move_to(x, y)
        # turn_angle = random.choice([0, 90, 180, 270]) # squares
        # turn_angle = random.choice([0, 60, 120]) # triangles
        turn_angle = random.choice(angles)
        left(turn_angle)

random_painting(100, 10, 3, 600, 600) 
```

```python
# user login system
users = {"michelle":"password", "liam":"67", "andrew":"43"}

# add users to the dictionary
def add_user(users):
    # use a while loop and get user input for username and password.
    # how do you avoid overwriting old users?
    adding  = True # flag variable
    while adding:
        new_username = input("Add username here:")
        # check if user exists
        if new_username in users:
            print("error: user already exists")
        else:
            new_password = input("add password here:")
            users[new_username] = new_password 
            print(f"User {new_username} added succesfully")
            adding = False
# add_user(users)
```

```python
def login(users):
    username = input("please enter your username:")
    password = input("please enter your password:")
    if username in users:
        if users[username] == password:
            print(f"Welcome, {username}")
            return True
        else:
            print("invalid password")
            return False
    else:
        print("invalid username")
        return False
# login(users)
```

```python
def chatbot():
    # store keywords and description
    commands = {"exit" : "Exits the chatbot", 
                "help" : "Shows available commands",
                "hello" : "say hello"}
    
    print("Hi! I'm a chatbot. Type 'help' for commands.\n")
    while True:
        user_input = input("You: ")
        if "help" in user_input:
            print("\nAvailable commands:")
            # loop through the commands dictionary
            for command, description in commands.items():
                print(f"- {command}: description")
        elif "hello" in user_input:
            print("hello!")
        elif "exit" in user_input:
            print("goodbye!")
            break
        else:
            print("sorry, I don't understand that command.\n")

chatbot()
```
