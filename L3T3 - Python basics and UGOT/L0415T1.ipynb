```python
from ugot import ugot
got = ugot.UGOT()
got.initialize("192.168.88.1")

got.show_light_rgb_effect(0, 0, 255, 0)

# forward
got.mecanum_move_speed_times(direction=0, speed=20, times=30, unit=1)
# backward
got.mecanum_move_speed_times(direction=1, speed=20, times=30, unit=1)
# turn left 90 degrees
got.mecanum_turn_speed_times(turn=2, speed=40, times=90, unit=2)
# turn right 90 degrees
got.mecanum_turn_speed_times(turn=3, speed=40, times=90, unit=2)
```

```python
# challenge: follow the line
got.mecanum_move_speed_times(direction=0, speed=20, times=60, unit=1)
got.mecanum_turn_speed_times(turn=2, speed=40, times=45, unit=2)
got.mecanum_move_speed_times(direction=0, speed=20, times=60, unit=1)
got.mecanum_turn_speed_times(turn=3, speed=40, times=45, unit=2)
got.mecanum_move_speed_times(direction=0, speed=20, times=60, unit=1)
```
