```python
from ugot import ugot
got = ugot.UGOT()
got.initialize("192.168.1.xxx")

got.load_models(["line_recognition"])
got.set_track_recognition_line(0)

while True:
    no_line_count = 0
    line_info = got.get_single_track_total_info()
    if line_info:
        offset = line_info[0]
        line_type = line_info[1]
        if line_type == 0:
            no_line_count += 1
            if no_line_count > 4:
                got.mecanum_translate_speed_times(0, 13, 10, 1)
                got.mecanum_stop()
        else:
            no_line_count = 0
        rot = int(offset*0.4)
        got.mecanum_move_xyz(0, 13, rot)
```
