```python
# comparison operators
print(8 > 7) # more than
print(8 > 10)

print(8 == 8) # equal
print("lucas" == "wilson")
name = "allen"
print(name == "allen")
print("winston" == "Winston") # False - case-sensitive

print(6 < 7) # less than

print("morgan" != "rylan") # NOT equal
print(8 >= 8) # more than OR equal
```
