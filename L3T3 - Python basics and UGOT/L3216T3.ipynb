# Session 7

REVIEW: using a while loop and conditions (if, elif, else), print numbers 1-40. 
- If it is divisible by 3, print "titan" instead of the number.
- If it is divisible by 5, print "tech" instead of the number.
- If it is divisible by both 3 and 5, print "titan tech" instead of the number.
- Otherwise, print the number.

Hint: in Python, the modulo operator `%` works like this: `a%b` gives you the remainder after dividing `a` by `b`. For example, `5%2` equals 1, `5%3` equals 2, `5%4` equals 1. What is the remainder if `a` is divisible by `b`?

```python
i = 1
while i < 41:
    if i % 3 == 0 and i % 5 == 0: 
        print("titan tech")
    elif i % 5 == 0:
        print("tech")
    elif i % 3 == 0: # remainder when dividing by 3 is 0, so i is divisible by 3
        print("titan")
    else:
        print(i)
    i += 1
```

```python
import random

num = random.randint(1, 100)
chances = 5
guessed = False # flag variable
while chances > 0:
    guess = int(input("guess number"))
    if guess == num:
        print("correct")
        guessed = True
        break # exit the loop
    elif guess > num:
        print("too high")
    else:
        print("too low")
    chances -= 1
if not guessed:
    print("out of chances")
```

```python
# number guessing game
from ugot import ugot
got = ugot.UGOT()
got.initialize("192.168.1.xxx")

import random
number = random.randint(1, 10)
chance = 0 # how many chances you have used so far - START AT ZERO
guessed = False # flag variable - keep track of whether you guessed the number

while chance < 5: # 5 chances to guess
    chance += 1   # add 1 to the chance variable
    guess = int(input("Guess a number (1-10): "))  # get user input and convert to integer
    dist = got.read_distance_data(21)
    random_distance = random.randint(5, 15)
    if guess == number:
        guessed = True
        if dist > 50: # if no obstacles in front of robot
            got.mecanum_move_speed_times(0,25,random_distance, 1) # move forward
            print(f"The robot just moved {random_distance} cm")
        break
    else:
        print("try again")

if guessed:
    print("You guessed it!")
else:
    print("You are out of chances 🥲😭")
```

```python
# number guessing game
from ugot import ugot
got = ugot.UGOT()
got.initialize("192.168.1.xxx")

import random
number = random.randint(1, 10)
chance = 0 # how many chances you have used so far - START AT ZERO
guessed = False # flag variable - keep track of whether you guessed the number

while chance < 5: # 5 chances to guess
    got.screen_display_background(0) 
    chance += 1   # add 1 to the chance variable
    guess = int(input("Guess a number (1-10): "))  # get user input and convert to integer
    dist = got.read_distance_data(21)
    random_distance = random.randint(5, 15)
    if guess == number:
        got.screen_display_background(6) 
        guessed = True
        if dist > 20: # if no obstacles in front of robot
            got.mecanum_move_speed_times(0,25,random_distance, 1) # move forward
            print(f"The robot just moved {random_distance} cm")
        break
    elif guess < number:
        got.screen_display_background(3)
        print(f"{guess} is too low")
    else: # don't need condition
        got.screen_display_background(3)
        print(f"{guess} is too high")

if guessed:
    print("You guessed it!")
else:
    print(f"You are out of chances 🥲😭 The number was {number}")
```
