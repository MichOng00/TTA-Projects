```python
import pygame
import random

pygame.init()
screen = pygame.display.set_mode((800, 600)) # (width, height)
pygame.display.set_caption("hello")

player_img = pygame.image.load("zelda.png")
player_img = pygame.transform.scale(player_img, (100, 100))
enemy_img = pygame.image.load("ganon.png")
enemy_img = pygame.transform.scale(enemy_img, (100, 100))

x = 100
y = 100
ex = 400
ey = 300

enemy_direction = [random.choice([-1, 1]), random.choice([-1, 1])]
enemy_speed = 1

font = pygame.font.SysFont(None, 30)

hits = 0

running = True
while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    keys = pygame.key.get_pressed()
    if keys[pygame.K_LEFT]:
        x -= 1
    if keys[pygame.K_RIGHT]:
        x += 1
    if keys[pygame.K_UP]:
        y -= 1
    if keys[pygame.K_DOWN]:
        y += 1

    if x < 0:
        x = 0
    elif x > 800 - 100:
        x = 800 - 100
    if y < 0:
        y = 0
    elif y > 600 - 100:
        y = 600 - 100

    # enemy movement
    ex += enemy_direction[0] * enemy_speed
    ey += enemy_direction[1] * enemy_speed
    # bounce off the window edges
    if ex <= 0 or ex >= 800 - 100:
        enemy_direction[0] *= -1
    if ey <= 0 or ey >= 600 - 100:
        enemy_direction[1] *= -1

    screen.fill((255, 255, 255))

    # draw shapes
    pygame.draw.circle(screen, (255, 0, 0), (400, 200), 50) # (surface, color, coordinates, radius)
    # pygame.draw.rect(screen, (0, 0, 255), (x, y, 100, 100)) # (x, y, width, height)
    player_rect = pygame.Rect(x, y, 100, 100)
    pygame.draw.rect(screen, (0, 255, 0), player_rect, width = 3)
    screen.blit(player_img, (x, y))

    enemy_rect = pygame.Rect(ex, ey, 100, 100)
    pygame.draw.rect(screen, (0, 255, 0), enemy_rect, width = 3)
    screen.blit(enemy_img, (ex, ey))

    # handle collision
    if player_rect.colliderect(enemy_rect):
        hits += 1
        screen.blit(font.render("You got hit!", True, (255, 0, 0)), (x, y))
        pygame.display.flip()
        pygame.time.wait(1000) # wait 1 second for you to see the text
        x = random.randint(0, 800)
        y = random.randint(0, 600)
    
    hits_text = font.render(f"Hits: {hits}", True, (0, 0, 0))
    screen.blit(hits_text, (10, 10))
    pygame.display.flip()

pygame.quit()
```
